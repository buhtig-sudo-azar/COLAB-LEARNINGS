## 1. Что такое высказывание?

### Жизненная аналогия: не любая фраза — высказывание

Скажи вслух:
- «Сократ — философ» — это **высказывание**. Можно проверить: да, истинно.
- «Земля плоская» — это **высказывание**. Можно проверить: нет, ложно.
- «Который час?» — это **не** высказывание. Это вопрос, к нему не применишь «истинно/ложно».
- «Иди сюда!» — это **не** высказывание. Это приказ.
- «Какая красивая музыка!» — это **не** высказывание. Это эмоция, нет объективной истинности.

### Определение

> **Высказывание** — это утверждение, о котором можно сказать, что оно **истинно** или **ложно**, но не то и другое одновременно.

**Три условия:**
1. Это **повествовательное** предложение (не вопрос, не приказ, не восклицание).
2. Можно сказать: **истинно (И)** или **ложно (Л)**.
3. Не может быть одновременно и тем, и другим.

### Обозначения истинности

| Русское | Английское | Python | Числовое |
|---|---|---|---|
| **И** (истина) | True / T | `True` | 1 |
| **Л** (ложь) | False / F | `False` | 0 |

### Пропозициональные переменные

Любое простое высказывание можно обозначить **буквой** (обычно большой латинской: $P, Q, R, \ldots$).

| Буква | Высказывание | Значение |
|---|---|---|
| $P$ | «Сократ — философ» | И (True) |
| $Q$ | «42 — целое число» | И (True) |
| $R$ | «Земля плоская» | Л (False) |

Зачем обозначать буквой? Чтобы коротко записывать составные конструкции: $P \land Q$, $\neg R$, $P \to Q$. Так же, как в алгебре: $x + y$ короче, чем «первое число плюс второе».

> **Зачем это в ML.** Любое правило классификации — это высказывание. «Если возраст > 18 И доход > 50к, то выдать кредит» — это логическая формула. Решающие деревья буквально строятся из таких условий.

## 2. Простые и составные высказывания

### Простые

**Простое высказывание** — это высказывание «как есть», без связок. Примеры:
- $P$ = «Идёт дождь»
- $Q$ = «Дует ветер»
- $R$ = «На улице +20°C»

### Составные

Из простых можно собирать **составные** с помощью **логических связок**:

| Связка (русск.) | Символ | Название | Пример |
|---|---|---|---|
| «не» | $\neg$ | отрицание | $\neg P$ = «не идёт дождь» |
| «и» | $\land$ | конъюнкция | $P \land Q$ = «дождь и ветер» |
| «или» | $\lor$ | дизъюнкция | $P \lor Q$ = «дождь или ветер» |
| «либо..., либо...» | $\oplus$ | XOR | $P \oplus Q$ = «либо дождь, либо ветер, но не оба сразу» |
| «если..., то...» | $\to$ | импликация | $P \to Q$ = «если дождь, то ветер» |

### Пример: переход к символьной форме

Пусть $P$ = «логика — забава», $Q$ = «пингвины розового цвета».

| Фраза на русском | Символьная форма |
|---|---|
| «Логика — не забава, и пингвины розового цвета» | $(\neg P) \land Q$ |
| «Пингвины не розового цвета, да и логика — не забава» | $(\neg Q) \land (\neg P)$ |
| «Либо логика — забава, либо пингвины розовые» | $P \oplus Q$ |
| «Если логика — забава, то пингвины розовые» | $P \to Q$ |

> **Важно:** в логике «и» и «или» **коммутативны**: $P \land Q \equiv Q \land P$. В естественном языке порядок иногда важен: «у Джейн родился ребёнок, и она вышла замуж» — не то же самое, что «Джейн вышла замуж, и у неё родился ребёнок» (хронология!). Но логику интересует только **истинность**, а не хронология.

## 3. Отрицание (НЕ, ¬)

### Жизненная аналогия: «зонтик не нужен»

Друг говорит: «На улице дождь» ($P$). Ты отвечаешь: «На улице НЕ дождь» ($\neg P$). Если дождь действительно идёт, то $P$ — истина, а $\neg P$ — ложь. Если дождя нет, наоборот.

### Определение

> **Отрицанием** высказывания $P$ называется высказывание $\neg P$, истинностное значение которого **строго противоположно** значению $P$.

Читается: **«не P»** или **«неверно, что P»**.

### Таблица истинности

| $P$ | $\neg P$ |
|---|---|
| И | Л |
| Л | И |

Всего 2 строки, потому что у $P$ есть 2 возможных значения.

### Обозначения

В разных учебниках:
- $\neg P$ (стандарт)
- $\overline{P}$ (как «чёрточка сверху», булева алгебра)
- $!P$ (в программировании: C, Java)
- `not P` (в Python)

### Двойное отрицание

$\neg(\neg P) \equiv P$ — «не не дождь» = «дождь». Логично!

> **Зачем это в ML.** Любая маска в NumPy — это отрицание. `~mask` даёт элементы, которые **не** попали в исходную маску. Это и есть $\neg$.

## 4. Конъюнкция (И, ∧)

### Жизненная аналогия: «билет в кино И попкорн»

Идёшь в кино. Хочешь: билет **И** попкорн. Если хоть одного нет — вечер испорчен. Только **оба** вместе делают тебя счастливым.

### Определение

> **Конъюнкцией** двух высказываний $P$ и $Q$ называется составное высказывание $P \land Q$, истинное **только когда истинны обе части**.

Читается: **«P и Q»**.

### Таблица истинности

| $P$ | $Q$ | $P \land Q$ |
|---|---|---|
| И | И | **И** |
| И | Л | Л |
| Л | И | Л |
| Л | Л | Л |

Из 4 строк только **одна** даёт И. Это и есть «оба одновременно».

### Примеры

- «2 + 2 = 4 И 3 + 3 = 6» → И (оба истинны)
- «2 + 2 = 4 И 3 + 3 = 7» → Л (второе ложно)
- «2 + 2 = 5 И 3 + 3 = 6» → Л (первое ложно)
- «2 + 2 = 5 И 3 + 3 = 7» → Л (оба ложны)

### Свойства

- **Коммутативность:** $P \land Q \equiv Q \land P$ (порядок не важен)
- **Ассоциативность:** $(P \land Q) \land R \equiv P \land (Q \land R)$
- **Идемпотентность:** $P \land P \equiv P$ («дождь и дождь» = «дождь»)
- **$P \land \neg P \equiv$ Л** (не может одновременно идти и не идти дождь)

> **Зачем это в ML.** Фильтр датасета `df[(df.age > 18) & (df.income > 50)]` — это конъюнкция двух предикатов. Только строки, где **оба** условия истинны, попадут в результат.

## 5. Дизъюнкция (ИЛИ, ∨)

### Жизненная аналогия: «чай ИЛИ кофе»

Тебя спрашивают: «Будешь чай или кофе?» Если ты согласишься на чай — да. На кофе — да. На оба (например, чай сейчас, кофе потом) — тоже да. Откажешься от обоих — нет.

### Определение

> **Дизъюнкцией** двух высказываний $P$ и $Q$ называется высказывание $P \lor Q$, истинное, если **хотя бы одна** из частей истинна.

Читается: **«P или Q»**.

### Таблица истинности

| $P$ | $Q$ | $P \lor Q$ |
|---|---|---|
| И | И | **И** |
| И | Л | **И** |
| Л | И | **И** |
| Л | Л | Л |

Из 4 строк **три** дают И. Только «оба ложны» → Л.

### «ИЛИ» в логике vs «ИЛИ» в жизни

В **логике** «или» — **нестрогое**: «ИЛИ то, ИЛИ другое, ИЛИ оба». Это дизъюнкция $\lor$.

В **жизни** «или» часто **строгое**: «ИЛИ то, ИЛИ другое, но НЕ оба». Это исключающее ИЛИ ($\oplus$, см. следующий раздел).

**Пример:** «Идёт дождь или снег». Дождь и снег могут идти одновременно. Поэтому связка нестрогая: $\text{дождь} \lor \text{снег}$.

**Контрпример:** «Сегодня понедельник или вторник». Не может быть и понедельник, и вторник одновременно. Поэтому связка строгая: $\text{пн} \oplus \text{вт}$.

### Свойства

- **Коммутативность:** $P \lor Q \equiv Q \lor P$
- **Ассоциативность:** $(P \lor Q) \lor R \equiv P \lor (Q \lor R)$
- **Идемпотентность:** $P \lor P \equiv P$
- **$P \lor \neg P \equiv$ И** (закон исключённого третьего: либо дождь, либо не дождь — третьего не дано)

> **Зачем это в ML.** Если в классификаторе логистическая регрессия даёт `class = 1 if (prob > 0.5) OR (some_rule)` — это дизъюнкция.

````python
# ============================================================
# Реализуем базовые логические операции и строим таблицы истинности
# ============================================================
# Каждая операция — это просто функция от булевых аргументов.

def neg(p):
    """Отрицание: НЕ p.
    Возвращает True, если p=False, и наоборот.
    """
    return not p                                 # в Python: not — это ¬

def conj(p, q):
    """Конъюнкция: p И q.
    Истинно ТОЛЬКО когда оба истинны.
    """
    return p and q                               # в Python: and — это ∧

def disj(p, q):
    """Дизъюнкция: p ИЛИ q.
    Истинно когда хотя бы одно истинно.
    """
    return p or q                                # в Python: or — это ∨

def xor(p, q):
    """Исключающее ИЛИ: либо p, либо q, но не оба.
    Истинно когда p и q РАЗНЫЕ.
    """
    return p != q                                # True, если значения не совпадают

def impl(p, q):
    """Импликация: если p, то q.
    Ложна ТОЛЬКО когда p=True, q=False. Во всех остальных случаях истинна.
    """
    return (not p) or q                          # p→q эквивалентно (¬p) ∨ q

# Демонстрируем все операции на одной таблице
print("Базовые операции над двумя переменными P, Q:")
print()
print(f"{'P':>5} {'Q':>5} | {'¬P':>5} {'P∧Q':>5} {'P∨Q':>5} {'P⊕Q':>5} {'P→Q':>5}")
print("-" * 50)

# itertools.product([True, False], repeat=2) даёт все 4 пары
for p, q in itertools.product([True, False], repeat=2):
    # Преобразуем True/False в T/F для красоты
    p_str = 'T' if p else 'F'
    q_str = 'T' if q else 'F'
    np_str = 'T' if neg(p) else 'F'
    pq_str = 'T' if conj(p, q) else 'F'
    por_str = 'T' if disj(p, q) else 'F'
    pxor_str = 'T' if xor(p, q) else 'F'
    pimp_str = 'T' if impl(p, q) else 'F'
    print(f"{p_str:>5} {q_str:>5} | {np_str:>5} {pq_str:>5} {por_str:>5} {pxor_str:>5} {pimp_str:>5}")

print()
print("Сверь с таблицами в разделах 3-7: всё должно совпадать!")
````

````python
# ============================================================
# Картинка 1: визуализация таблиц истинности базовых операций
# ============================================================
# Рисуем 5 операций: НЕ, И, ИЛИ, XOR, →
# Истинные ячейки — зелёные, ложные — красные

# Все комбинации (P, Q) — 4 штуки
combos = list(itertools.product([True, False], repeat=2))
# Для НЕ нужен только P, но для согласованности берём ту же сетку

operations = [
    ('¬P  (НЕ)',          lambda p, q: neg(p),       ['#e74c3c', '#27ae60']),
    ('P ∧ Q  (И)',        lambda p, q: conj(p, q),   ['#e74c3c', '#e74c3c', '#e74c3c', '#27ae60']),
    ('P ∨ Q  (ИЛИ)',      lambda p, q: disj(p, q),   ['#e74c3c', '#27ae60', '#27ae60', '#27ae60']),
    ('P ⊕ Q  (XOR)',      lambda p, q: xor(p, q),    ['#e74c3c', '#27ae60', '#27ae60', '#e74c3c']),
    ('P → Q  (импликация)', lambda p, q: impl(p, q), ['#e74c3c', '#27ae60', '#27ae60', '#27ae60']),
]

fig, axes = plt.subplots(1, 5, figsize=(16, 4))

for ax, (title, op, colors) in zip(axes, operations):
    # 4 ячейки 2x2
    for idx, (p, q) in enumerate(combos):
        row = idx // 2     # 0 или 1 (P)
        col = idx % 2      # 0 или 1 (Q)
        val = op(p, q)
        color = '#27ae60' if val else '#e74c3c'
        rect = Rectangle((col, 1 - row), 1, 1,
                          facecolor=color, alpha=0.75,
                          edgecolor='#2c3e50', linewidth=2)
        ax.add_patch(rect)
        # Текст внутри ячейки: результат
        ax.text(col + 0.5, 1 - row + 0.5, 'T' if val else 'F',
                ha='center', va='center', fontsize=22,
                color='white', fontweight='bold')
        # Подпись строки/колонки
        ax.text(col + 0.5, 2.15, f"Q={'T' if q else 'F'}",
                ha='center', fontsize=10, color='#2c3e50')
        ax.text(-0.25, 1 - row + 0.5, f"P={'T' if p else 'F'}",
                ha='right', va='center', fontsize=10, color='#2c3e50')

    ax.set_xlim(-0.5, 2.2)
    ax.set_ylim(-0.3, 2.5)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title(title, fontsize=12, fontweight='bold', pad=10)

plt.suptitle('Таблицы истинности базовых операций\nЗелёный = Истина, Красный = Ложь',
             fontsize=14, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()
````

## 6. Исключающее ИЛИ (XOR, ⊕)

### Жизненная аналогия: «или пан, или пропал»

Ты на развилке: «Либо идёшь направо, либо налево». Не можешь пойти в обе стороны одновременно. Это **строгое** «или».

### Определение

> **Исключающим ИЛИ** (XOR) двух высказываний $P$ и $Q$ называется высказывание $P \oplus Q$, истинное **тогда и только тогда, когда значения $P$ и $Q$ не совпадают**.

Читается: **«либо P, либо Q (но не оба)»**.

### Таблица истинности

| $P$ | $Q$ | $P \oplus Q$ |
|---|---|---|
| И | И | Л |
| И | Л | **И** |
| Л | И | **И** |
| Л | Л | Л |

Из 4 строк **две** дают И — это случаи, когда $P \neq Q$.

### Сравнение: ИЛИ vs XOR

| $P$ | $Q$ | $P \lor Q$ (ИЛИ) | $P \oplus Q$ (XOR) |
|---|---|---|---|
| И | И | И | Л |
| И | Л | И | И |
| Л | И | И | И |
| Л | Л | Л | Л |

Разница только в первой строке: когда **оба истинны**. ИЛИ говорит «да», XOR говорит «нет».

### Где встречается

- **Дни недели:** «Сегодня пн или вт» → XOR (не может быть оба сразу).
- **Выбор в кафе:** «Кофе или чай» → обычно XOR (если ты не попросишь полпорции каждого).
- **Погода:** «Дождь или снег» → обычно ИЛИ (могут идти одновременно).
- **Битовый XOR в криптографии:** основа шифрования — если $P$ — бит сообщения, $Q$ — бит ключа, то $P \oplus Q$ — зашифрованный бит. Повторное XOR с тем же ключом восстанавливает $P$.

### Свойства

- **Коммутативность:** $P \oplus Q \equiv Q \oplus P$
- **Ассоциативность:** $(P \oplus Q) \oplus R \equiv P \oplus (Q \oplus R)$
- **$P \oplus P \equiv$ Л** (XOR с самим собой = ложь)
- **$P \oplus F \equiv P$** (XOR с ложью ничего не меняет)

> **Зачем это в ML.** В neural networks функция активации step иногда сравнивает знаки — это XOR. В контрольных суммах (CRC, checksum) — тоже XOR. В классификаторах с двумя классами «угадал/не угадал» — это XOR между предсказанием и истиной.

## 7. Импликация (если → то, →)

### Жизненная аналогия: обещание

Ты обещаешь: «**Если** завтра будет суббота, **то** я приду на встречу».

Давай разберём 4 случая:
1. Завтра суббота, ты пришёл → обещание **сдержано** (истинно).
2. Завтра суббота, ты **не** пришёл → обещание **нарушено** (ложно).
3. Завтра **не суббота**, ты пришёл → обещание **не нарушено** (истинно — ты же не обещал не приходить в другие дни!).
4. Завтра **не суббота**, ты **не** пришёл → обещание **не нарушено** (истинно).

Только один случай даёт ложь: посылка истинна, а заключение ложно.

### Определение

> **Импликацией** двух высказываний $P$ и $Q$ называется высказывание $P \to Q$, **ложное тогда и только тогда, когда $P$ истинно, а $Q$ ложно**. Во всех остальных случаях истинно.

- $P$ — **посылка** (антецедент)
- $Q$ — **заключение** (консеквент)

Читается: **«если P, то Q»** или **«из P следует Q»** или **«P влечёт Q»**.

### Таблица истинности

| $P$ | $Q$ | $P \to Q$ |
|---|---|---|
| И | И | **И** |
| И | Л | Л |
| Л | И | **И** |
| Л | Л | **И** |

### Разбираем парадокс

Многих смущает: почему $F \to T$ и $F \to F$ — оба истинны?

**Пример.** Пусть $P$ = «1 = 5» (ложно), $Q$ = «3 = 7» (ложно), $R$ = «4 = 4» (истинно).

- $P \to Q$: «если 1=5, то 3=7». Прибавив 2 к обеим частям равенства 1=5, получим 3=7. Значит, из ложной посылки логически следует ложное заключение — импликация **истинна**.
- $P \to R$: «если 1=5, то 4=4». Если 1=5, вычтем 3: -2=2. Возведём в квадрат: 4=4. Из ложной посылки получили истинное заключение — импликация **истинна**.

**Логическое правило:** «из лжи следует что угодно». Это не баг, а фича: ложная посылка не может «нарушить» обещание.

### Эквивалентность

$P \to Q \equiv \neg P \lor Q$

То есть «если P, то Q» = «либо не P, либо Q».

**Проверка:** «если завтра суббота, то я приду» = «либо завтра не суббота, либо я приду». Да, логично!

### Отрицание импликации

$\neg(P \to Q) \equiv P \land \neg Q$

Читается: импликация **ложна** тогда и только тогда, когда посылка истинна, а заключение ложно.

> **Зачем это в ML.** Любая условная конструкция `if condition: action` — это импликация. Градиентный спуск использует правило «если loss растёт, то уменьши learning rate» — это импликация. Логистическая регрессия: «если взвешенная сумма > 0, то класс 1» — тоже импликация.

````python
# ============================================================
# Импликация: проверка свойств через таблицу истинности
# ============================================================
# Проверяем, что p→q ≡ (¬p) ∨ q

print("Проверка: p→q ≡ (¬p) ∨ q")
print()
print(f"{'P':>5} {'Q':>5} | {'P→Q':>6} {'¬P':>5} {'¬P∨Q':>6} | {'Совпадает?':>11}")
print("-" * 55)

all_match = True
for p, q in itertools.product([True, False], repeat=2):
    left = impl(p, q)                         # p → q
    right = disj(neg(p), q)                   # (¬p) ∨ q
    match = left == right
    if not match:
        all_match = False
    p_str = 'T' if p else 'F'
    q_str = 'T' if q else 'F'
    left_str = 'T' if left else 'F'
    np_str = 'T' if neg(p) else 'F'
    right_str = 'T' if right else 'F'
    match_str = '✓ да' if match else '✗ нет'
    print(f"{p_str:>5} {q_str:>5} | {left_str:>6} {np_str:>5} {right_str:>6} | {match_str:>11}")

print()
print(f"Все 4 строки совпадают: {'ДА' if all_match else 'НЕТ'}")
print(f"Значит, p→q ≡ (¬p) ∨ q — доказано.")

# Проверка отрицания импликации: ¬(p→q) ≡ p ∧ ¬q
print()
print("=" * 55)
print("Проверка: ¬(p→q) ≡ p ∧ ¬q")
print()
print(f"{'P':>5} {'Q':>5} | {'¬(P→Q)':>8} {'P':>5} {'¬Q':>5} {'P∧¬Q':>6} | {'Совпадает?':>11}")
print("-" * 60)

all_match2 = True
for p, q in itertools.product([True, False], repeat=2):
    left = neg(impl(p, q))                    # ¬(p → q)
    right = conj(p, neg(q))                   # p ∧ ¬q
    match = left == right
    if not match:
        all_match2 = False
    p_str = 'T' if p else 'F'
    q_str = 'T' if q else 'F'
    left_str = 'T' if left else 'F'
    nq_str = 'T' if neg(q) else 'F'
    right_str = 'T' if right else 'F'
    match_str = '✓ да' if match else '✗ нет'
    print(f"{p_str:>5} {q_str:>5} | {left_str:>8} {p_str:>5} {nq_str:>5} {right_str:>6} | {match_str:>11}")

print()
print(f"Все 4 строки совпадают: {'ДА' if all_match2 else 'НЕТ'}")
print(f"Значит, ¬(p→q) ≡ p ∧ ¬q — доказано. Импликация ложна ТОЛЬКО когда p=True, q=False.")
````

## 8. Эквивалентность и законы де Моргана

### Что значит «эквивалентны»

Два составных высказывания **логически эквивалентны** (обозначается $\equiv$), если они **принимают одинаковые значения истинности на любом наборе переменных**.

Проверка: строим таблицы истинности обоих и сравниваем столбцы. Если полностью совпадают — эквивалентны.

### Законы де Моргана — самые важные

**Закон 1:** $\neg(P \land Q) \equiv \neg P \lor \neg Q$

Читается: «НЕ (P И Q) = (НЕ P) ИЛИ (НЕ Q)».

**Жизненный пример.** «НЕ (идёт дождь И идёт снег)» = «НЕ дождь ИЛИ НЕ снег». Действительно: чтобы не было «дождя со снегом», достаточно, чтобы хотя бы одного из них не было.

**Закон 2:** $\neg(P \lor Q) \equiv \neg P \land \neg Q$

Читается: «НЕ (P ИЛИ Q) = (НЕ P) И (НЕ Q)».

**Жизненный пример.** «НЕ (дождь ИЛИ снег)» = «НЕ дождь И НЕ снег». Чтобы не было ни дождя, ни снега, нужно чтобы **и** дождя не было, **и** снега не было.

### Обобщение на n переменных

- $\neg(P_1 \land P_2 \land \ldots \land P_n) \equiv \neg P_1 \lor \neg P_2 \lor \ldots \lor \neg P_n$
- $\neg(P_1 \lor P_2 \lor \ldots \lor P_n) \equiv \neg P_1 \land \neg P_2 \land \ldots \land \neg P_n$

### Другие важные эквивалентности

| Название | Формула |
|---|---|
| Двойное отрицание | $\neg(\neg P) \equiv P$ |
| Коммутативность И | $P \land Q \equiv Q \land P$ |
| Коммутативность ИЛИ | $P \lor Q \equiv Q \lor P$ |
| Ассоциативность И | $(P \land Q) \land R \equiv P \land (Q \land R)$ |
| Дистрибутивность | $P \land (Q \lor R) \equiv (P \land Q) \lor (P \land R)$ |
| Идемпотентность | $P \land P \equiv P$ |
| Законы поглощения | $P \land (P \lor Q) \equiv P$ |
| Закон исключённого третьего | $P \lor \neg P \equiv$ И |
| Закон противоречия | $P \land \neg P \equiv$ Л |
| Импликация через ИЛИ | $P \to Q \equiv \neg P \lor Q$ |
| Отрицание импликации | $\neg(P \to Q) \equiv P \land \neg Q$ |

> **Зачем это в ML.** Когда ты фильтруешь датасет «НЕ (класс = кошка ИЛИ класс = собака)», по законам де Моргана это «класс ≠ кошка И класс ≠ собака». Так код работает быстрее — одно условие instead двух OR.

````python
# ============================================================
# Проверка законов де Моргана через таблицы истинности
# ============================================================

print("Закон 1: ¬(P ∧ Q) ≡ ¬P ∨ ¬Q")
print()
print(f"{'P':>5} {'Q':>5} | {'¬(P∧Q)':>8} {'¬P':>5} {'¬Q':>5} {'¬P∨¬Q':>7} | {'Совпадает?':>11}")
print("-" * 60)

all_match1 = True
for p, q in itertools.product([True, False], repeat=2):
    left = neg(conj(p, q))                    # ¬(P ∧ Q)
    right = disj(neg(p), neg(q))              # ¬P ∨ ¬Q
    match = left == right
    if not match:
        all_match1 = False
    p_str = 'T' if p else 'F'
    q_str = 'T' if q else 'F'
    left_str = 'T' if left else 'F'
    np_str = 'T' if neg(p) else 'F'
    nq_str = 'T' if neg(q) else 'F'
    right_str = 'T' if right else 'F'
    match_str = '✓ да' if match else '✗ нет'
    print(f"{p_str:>5} {q_str:>5} | {left_str:>8} {np_str:>5} {nq_str:>5} {right_str:>7} | {match_str:>11}")

print()
print(f"Закон 1 подтверждён: {'ДА' if all_match1 else 'НЕТ'}")

print()
print("=" * 60)
print("Закон 2: ¬(P ∨ Q) ≡ ¬P ∧ ¬Q")
print()
print(f"{'P':>5} {'Q':>5} | {'¬(P∨Q)':>8} {'¬P':>5} {'¬Q':>5} {'¬P∧¬Q':>7} | {'Совпадает?':>11}")
print("-" * 60)

all_match2 = True
for p, q in itertools.product([True, False], repeat=2):
    left = neg(disj(p, q))                    # ¬(P ∨ Q)
    right = conj(neg(p), neg(q))              # ¬P ∧ ¬Q
    match = left == right
    if not match:
        all_match2 = False
    p_str = 'T' if p else 'F'
    q_str = 'T' if q else 'F'
    left_str = 'T' if left else 'F'
    np_str = 'T' if neg(p) else 'F'
    nq_str = 'T' if neg(q) else 'F'
    right_str = 'T' if right else 'F'
    match_str = '✓ да' if match else '✗ нет'
    print(f"{p_str:>5} {q_str:>5} | {left_str:>8} {np_str:>5} {nq_str:>5} {right_str:>7} | {match_str:>11}")

print()
print(f"Закон 2 подтверждён: {'ДА' if all_match2 else 'НЕТ'}")

print()
print("=" * 60)
print("ОБОБЩЕНИЕ: проверим ¬(P ∧ Q ∧ R) ≡ ¬P ∨ ¬Q ∨ ¬R на 8 комбинациях")
print()
all_match3 = True
count = 0
for p, q, r in itertools.product([True, False], repeat=3):
    left = not (p and q and r)                # ¬(P ∧ Q ∧ R)
    right = (not p) or (not q) or (not r)     # ¬P ∨ ¬Q ∨ ¬R
    match = left == right
    if not match:
        all_match3 = False
    count += 1
print(f"Проверено {count} комбинаций. Все совпадают: {'ДА' if all_match3 else 'НЕТ'}")
````

## 9. Предикаты

### Жизненная аналогия: шаблон с дыркой

Загадка: «$x$ — целое число, для которого $x^2 = x$». Это не высказывание — в нём есть переменная $x$, и мы не знаем, истина или ложь, пока не подставим значение.

- Подставим $x = 0$: $0^2 = 0$ — И.
- Подставим $x = 1$: $1^2 = 1$ — И.
- Подставим $x = 2$: $4 = 2$ — Л.

Это **предикат** — высказывание с дыркой.

### Определение

> **Предикат** — это высказывание, содержащее одну или несколько переменных, и принимающее значение истинности или ложности **в зависимости от значений этих переменных**.

**$n$-местный предикат** — функция:
$$P(x_1, x_2, \ldots, x_n) : M^n \to \{T, F\}$$

| Часть | Что значит |
|---|---|
| $P$ | имя предиката |
| $(x_1, \ldots, x_n)$ | переменные (аргументы) |
| $M^n$ | множество, из которого берутся аргументы |
| $\{T, F\}$ | множество значений: истина или ложь |

### Примеры

| Предикат | При каких $x$ истинен |
|---|---|
| $P(x): x^2 = x$ | $x = 0$ или $x = 1$ |
| $Q(x): x > 0$ | $x \in (0, +\infty)$ |
| $R(x, y): x + y = 0$ | когда $y = -x$ |
| $D(x): x \text{ делится на } 2$ | чётные числа |
| $E(x): x \text{ делится на } 3$ | числа, кратные 3 |

### Логические операции над предикатами

К предикатам применимы те же операции, что и к высказываниям: $\neg, \land, \lor, \oplus, \to$. Результат — новый предикат.

**Пример.** Если $D(x)$ = «$x$ делится на 2», $E(x)$ = «$x$ делится на 3», то:
- $D(x) \land E(x)$ — «$x$ делится и на 2, и на 3» = «делится на 6».
- $D(x) \lor E(x)$ — «делится на 2 или на 3».
- $\neg D(x)$ — «не делится на 2» = «нечётное».

> **Зачем это в ML.** Любая функция-фильтр `lambda x: x > 0` — это предикат. Лямбды в feature engineering — предикаты. Условие в `if` — предикат. Без предикатов нет ML.

````python
# ============================================================
# Предикаты: функции, возвращающие True/False
# ============================================================

# Предикат P(x): x² = x  (истинен для x=0 и x=1)
def P(x):
    """Предикат: x² = x."""
    return x * x == x                          # сравниваем x² и x

# Предикат D(x): x делится на 2
def D(x):
    """Предикат: x чётное (делится на 2)."""
    return x % 2 == 0                          # остаток от деления на 2 равен 0

# Предикат E(x): x делится на 3
def E(x):
    """Предикат: x делится на 3."""
    return x % 3 == 0                          # остаток от деления на 3 равен 0

# Предикат Q(x): x > 0
def Q(x):
    """Предикат: x положительное."""
    return x > 0

# Проверяем предикат P(x) на разных значениях
print("Предикат P(x): x² = x")
print(f"{'x':>5} | {'P(x)':>6}")
print("-" * 15)
for x in [-2, -1, 0, 1, 2, 3, 5, 10]:
    val = P(x)
    print(f"{x:>5} | {'T' if val else 'F':>6}")
print("Видно: P истинен только для x=0 и x=1.")

# Проверяем D, E и их комбинации
print()
print("=" * 50)
print("Предикаты D (делится на 2), E (делится на 3) и их комбинации")
print()
print(f"{'x':>3} | {'D':>3} {'E':>3} {'D∧E':>4} {'D∨E':>4} {'¬D':>3}")
print("-" * 28)
for x in range(0, 13):
    d = D(x)                                   # делится на 2?
    e = E(x)                                   # делится на 3?
    d_and_e = conj(d, e)                       # делится и на 2, и на 3 (т.е. на 6)
    d_or_e = disj(d, e)                        # делится хотя бы на одно
    not_d = neg(d)                             # нечётное
    print(f"{x:>3} | {'T' if d else 'F':>3} {'T' if e else 'F':>3} "
          f"{'T' if d_and_e else 'F':>4} {'T' if d_or_e else 'F':>4} {'T' if not_d else 'F':>3}")
````

## 10. Квантор всеобщности (∀)

### Жизненная аналогия: «все ученики»

Учитель говорит: «**Все** ученики 5 «А» сдали экзамен». Чтобы проверить, нужно опросить каждого. Если хотя бы один не сдал — утверждение ложно.

### Определение

> **Квантор всеобщности** $\forall$ (читается «для всех») превращает предикат $P(x)$ в высказывание $\forall x \, P(x)$, истинное, когда $P(x)$ истинно **для каждого** элемента $x$ из множества $M$.

| Часть | Что значит |
|---|---|
| $\forall$ | «для всех», «для каждого» |
| $x$ | переменная, которая «связывается» квантором |
| $P(x)$ | предикат (условие) |

**Примеры:**
- $\forall x \in \mathbb{R}: x^2 \geq 0$ — «для любого вещественного $x$ его квадрат неотрицателен». **Истинно.**
- $\forall x \in \mathbb{Z}: x^2 = x$ — «для любого целого $x$ выполнено $x^2 = x$». **Ложно** (при $x=2$ это неверно).
- $\forall x \in \{2, 4, 6, 8\}: x \text{ чётное}$ — **истинно**.

### В Python

Аналог $\forall$ — функция `all()`:

```python
all(P(x) for x in M)  # True, если P(x) истинно для всех x из M
```

**Пример:**
```python
M = [2, 4, 6, 8]
all(x % 2 == 0 for x in M)   # True — все чётные
```

### Когда $\forall$ истинен, а когда ложен

| Ситуация | $\forall x \, P(x)$ |
|---|---|
| $P(x)$ истинно для всех $x$ из $M$ | **И** |
| Найдётся хотя бы один $x$, для которого $P(x)$ ложно | **Л** |

То есть достаточно **одного контрпримера**, чтобы опровергнуть $\forall$.

> **Зачем это в ML.** Утверждение «модель правильно классифицирует все объекты обучающей выборки» — это $\forall$. На практике такого почти не бывает, поэтому в ML чаще используют «большинство» или «в среднем».

## 11. Квантор существования (∃)

### Жизненная аналогия: «хотя бы один»

Сказать «**существует** хотя бы один ученик, который получил 5» — не нужно опрашивать всех. Достаточно найти одного отличника.

### Определение

> **Квантор существования** $\exists$ (читается «существует») превращает предикат $P(x)$ в высказывание $\exists x \, P(x)$, истинное, когда **существует** элемент $x \in M$, для которого $P(x)$ истинно.

| Часть | Что значит |
|---|---|
| $\exists$ | «существует», «найдётся» |
| $x$ | переменная |
| $P(x)$ | предикат |

**Примеры:**
- $\exists x \in \mathbb{Z}: x^2 = 4$ — «существует целое $x$ с $x^2 = 4$». **Истинно** ($x = 2$ или $x = -2$).
- $\exists x \in \mathbb{Z}: x^2 = 2$ — «существует целое $x$ с $x^2 = 2$». **Ложно** (таких целых нет).
- $\exists x \in \mathbb{R}: x^2 = 2$ — **истинно** ($x = \sqrt{2}$, оно вещественное).

### В Python

Аналог $\exists$ — функция `any()`:

```python
any(P(x) for x in M)  # True, если хотя бы для одного x из M предикат истинен
```

**Пример:**
```python
M = [1, 3, 5, 7]
any(x % 2 == 0 for x in M)   # False — нет ни одного чётного
```

### Сравнение $\forall$ и $\exists$

| | $\forall x \, P(x)$ | $\exists x \, P(x)$ |
|---|---|---|
| Смысл | «для всех» | «существует» |
| Python | `all(...)` | `any(...)` |
| Чтобы доказать истинность | проверить все | найти один пример |
| Чтобы доказать ложность | найти контрпример | проверить все |

> **Зачем это в ML.** «Существует ли конфигурация гиперпараметров с accuracy > 95%?» — это $\exists$. «Все ли предсказания на тесте верны?» — это $\forall$. Первое проще проверить (нашёл — доказано), второе сложнее.

````python
# ============================================================
# Кванторы ∀ (all) и ∃ (any) на списках чисел
# ============================================================

# Множество, на котором проверяем предикаты
M = list(range(0, 10))                        # [0, 1, 2, ..., 9]

print(f"Множество M = {M}")
print()

# --- Квантор существования ∃: существует ли x с нужным свойством? ---
# ∃x: D(x) ∧ E(x) — существует ли число, делящееся и на 2, и на 3 (т.е. на 6)?
exists_div_both = any(conj(D(x), E(x)) for x in M)
print(f"∃x (делится на 2 И на 3) на M: {'T' if exists_div_both else 'F'}")
print(f"  (ответ T, потому что 6 делится и на 2, и на 3)")

# ∃x: P(x) — существует ли x с x² = x?
exists_P = any(P(x) for x in M)
print(f"∃x (x² = x) на M: {'T' if exists_P else 'F'}")
print(f"  (ответ T, потому что 0²=0 и 1²=1)")

# ∃x: x² = 2 на целых числах
exists_x2_2 = any(x * x == 2 for x in range(-100, 101))
print(f"∃x ∈ ℤ (x² = 2): {'T' if exists_x2_2 else 'F'}")
print(f"  (ответ F — корень из 2 не целое число)")

print()

# --- Квантор всеобщности ∀: для всех x выполнено? ---
# ∀x: D(x) ∧ E(x) — все числа делятся и на 2, и на 3?
all_div_both = all(conj(D(x), E(x)) for x in M)
print(f"∀x (делится на 2 И на 3) на M: {'T' if all_div_both else 'F'}")
print(f"  (ответ F, потому что 1 не делится ни на 2, ни на 3)")

# ∀x: x² ≥ 0 — для всех x из M квадрат неотрицателен?
all_nonneg_square = all(x * x >= 0 for x in M)
print(f"∀x (x² ≥ 0) на M: {'T' if all_nonneg_square else 'F'}")
print(f"  (ответ T — квадрат любого целого неотрицателен)")

# ∀x: x > 0 — все числа положительные?
all_positive = all(x > 0 for x in M)
print(f"∀x (x > 0) на M: {'T' if all_positive else 'F'}")
print(f"  (ответ F, потому что 0 не положительный)")

print()
print("=" * 50)
print("ИТОГО:")
print(f"  ∃x (D ∧ E) = {'T' if exists_div_both else 'F'},   ∀x (D ∧ E) = {'T' if all_div_both else 'F'}")
print(f"  ∃x (x²=x) = {'T' if exists_P else 'F'},   ∀x (x²=x) = {'T' if any(P(x) for x in M) and all(P(x) for x in M) else 'F'}")
````

## 12. Отрицание кванторов

### Жизненная аналогия: «не все» = «есть хотя бы один плохой»

Сказать «**НЕ** все ученики сдали экзамен» — то же самое, что сказать «**существует** хотя бы один ученик, который **не** сдал».

И наоборот: «**НЕ** существует ни одного ученика, кто не сдал» = «**Все** ученики сдали».

### Законы отрицания кванторов

**Закон 1:** $\neg(\forall x \, P(x)) \equiv \exists x \, (\neg P(x))$

Читается: «НЕ (для всех x: P) = существует x, для которого не P».

**Закон 2:** $\neg(\exists x \, P(x)) \equiv \forall x \, (\neg P(x))$

Читается: «НЕ (существует x: P) = для всех x: не P».

### Запоминание

При отрицании квантор **меняется на парный**:
- $\forall \to \exists$
- $\exists \to \forall$

А предикат под ним **тоже отрицается**.

### Примеры

| Исходное | Отрицание |
|---|---|
| $\forall x: x > 0$ | $\exists x: x \leq 0$ |
| $\exists x: x^2 = 2$ | $\forall x: x^2 \neq 2$ |
| $\forall x: P(x) \to Q(x)$ | $\exists x: P(x) \land \neg Q(x)$ |

### В Python

```python
# ¬(∀x: P(x))  ≡  ∃x: ¬P(x)
not all(P(x) for x in M)  ==  any(not P(x) for x in M)

# ¬(∃x: P(x))  ≡  ∀x: ¬P(x)
not any(P(x) for x in M)  ==  all(not P(x) for x in M)
```

> **Зачем это в ML.** Часто проще доказать «существует ошибка», чем «все правильно». Законы отрицания кванторов позволяют переформулировать задачу: вместо «модель не делает ошибок» (трудно проверить) — «нет ни одного примера с ошибкой» (тоже трудно), но эквивалентно «для всех примеров модель верна». В зависимости от контекста одна форма удобнее другой.

````python
# ============================================================
# Проверка законов отрицания кванторов
# ============================================================

M = list(range(0, 10))                        # [0, 1, ..., 9]

print("Закон 1: ¬(∀x P(x)) ≡ ∃x ¬P(x)")
print()

# Проверяем на нескольких предикатах
test_predicates = [
    ('x > 0',      lambda x: x > 0),           # не все положительные (0 не положительный)
    ('x² ≥ 0',     lambda x: x * x >= 0),      # все квадраты неотрицательные
    ('x > 100',    lambda x: x > 100),         # никто не больше 100
    ('x - целое',  lambda x: isinstance(x, int)),  # все целые
]

print(f"{'Предикат P(x)':<20} | {'∀x P(x)':>10} {'¬(∀x P(x))':>12} | {'∃x ¬P(x)':>10} {'Совпадает?':>11}")
print("-" * 75)

for name, pred in test_predicates:
    all_P = all(pred(x) for x in M)            # ∀x P(x)
    not_all_P = not all_P                      # ¬(∀x P(x))
    exists_not_P = any(not pred(x) for x in M) # ∃x ¬P(x)
    match = not_all_P == exists_not_P
    match_str = '✓ да' if match else '✗ нет'
    print(f"{name:<20} | {'T' if all_P else 'F':>10} {'T' if not_all_P else 'F':>12} | "
          f"{'T' if exists_not_P else 'F':>10} {match_str:>11}")

print()
print("=" * 75)
print("Закон 2: ¬(∃x P(x)) ≡ ∀x ¬P(x)")
print()

print(f"{'Предикат P(x)':<20} | {'∃x P(x)':>10} {'¬(∃x P(x))':>12} | {'∀x ¬P(x)':>10} {'Совпадает?':>11}")
print("-" * 75)

for name, pred in test_predicates:
    exists_P = any(pred(x) for x in M)         # ∃x P(x)
    not_exists_P = not exists_P                # ¬(∃x P(x))
    all_not_P = all(not pred(x) for x in M)    # ∀x ¬P(x)
    match = not_exists_P == all_not_P
    match_str = '✓ да' if match else '✗ нет'
    print(f"{name:<20} | {'T' if exists_P else 'F':>10} {'T' if not_exists_P else 'F':>12} | "
          f"{'T' if all_not_P else 'F':>10} {match_str:>11}")

print()
print("Профессор: видишь, как красиво? Отрицание квантора = смена квантора + отрицание предиката.")
````

## 13. Несколько кванторов подряд

### Жизненная аналогия: «для любого x найдётся y»

Пусть $x, y$ — вещественные числа, $P(x, y)$ = «$x + y = 0$».

**Утверждение 1:** $\forall x \, \exists y \, P(x, y)$ — «для любого $x$ найдётся $y$ такой, что $x + y = 0$».

Это **истинно**: для любого $x$ берём $y = -x$, и сумма равна 0. Здесь $y$ **зависит от $x$** — для каждого $x$ свой $y$.

**Утверждение 2:** $\exists y \, \forall x \, P(x, y)$ — «существует такой $y$, что для всех $x$ выполнено $x + y = 0$».

Это **ложно**: нет такого «универсального» $y$, который компенсировал бы любой $x$. Если $y = -5$, то только для $x = 5$ сумма равна 0.

### Порядок кванторов ВАЖЕН

$\forall x \, \exists y \, P(x, y) \not\equiv \exists y \, \forall x \, P(x, y)$

**Правило:** первый квантор «связывает» внешнюю переменную, второй — внутреннюю. Если $\exists$ стоит **после** $\forall$, то «существующий» $y$ может **зависеть** от $x$. Если $\exists$ стоит **до** $\forall$, то $y$ должен быть **одним и тем же** для всех $x$.

### Пример: предел последовательности

Определение предела $\lim_{n \to \infty} x_n = a$:

$$\forall \varepsilon > 0 \, \exists M \in \mathbb{N} : \, \forall n > M \, |x_n - a| < \varepsilon$$

**Читается по частям:**
1. $\forall \varepsilon > 0$ — «для любого положительного эпсилон» (какого угодно, даже крошечного).
2. $\exists M \in \mathbb{N}$ — «существует натуральное число $M$» (зависит от $\varepsilon$!).
3. $\forall n > M$ — «для всех $n$ больше $M$».
4. $|x_n - a| < \varepsilon$ — «расстояние от $x_n$ до $a$ меньше $\varepsilon$».

**Смысл:** какую бы маленькую точность $\varepsilon$ ты ни задал, найдётся номер $M$, начиная с которого все члены последовательности лежат в $\varepsilon$-окрестности $a$.

**Порядок кванторов критичен:** $M$ выбирается **после** $\varepsilon$ — для каждого $\varepsilon$ своё $M$. Если поменять порядок на $\exists M \, \forall \varepsilon$, получится бессмыслица: «существует $M$, после которого все $x_n$ лежат в любой окрестности» — это верно только для постоянной последовательности.

### В Python: несколько кванторов

```python
# ∀x ∃y: x + y == 0 на R
all(any(x + y == 0 for y in M) for x in M)

# ∃y ∀x: x + y == 0 на R (должно быть False)
any(all(x + y == 0 for x in M) for y in M)
```

> **Зачем это в ML.** Определение сходимости градиентного спуска: «для любого $\varepsilon$ найдётся число шагов $M$, после которого loss отличается от минимума меньше чем на $\varepsilon$». Это формально записывается через $\forall \varepsilon \, \exists M$ — тот же шаблон, что и в определении предела.

````python
# ============================================================
# Несколько кванторов: ∀x ∃y P(x,y) vs ∃y ∀x P(x,y)
# ============================================================

# Множество, на котором работаем
M = list(range(-5, 6))                        # [-5, -4, ..., 5]

# Предикат P(x, y): x + y == 0
def P(x, y):
    """Предикат: x + y = 0 (т.е. y = -x)."""
    return x + y == 0

# Утверждение 1: ∀x ∃y P(x,y) — для любого x найдётся y с x+y=0
# Для каждого x проверяем, есть ли y такой, что P истинно
all_x_exists_y = all(
    any(P(x, y) for y in M)                    # ∃y: P(x, y)
    for x in M                                  # ∀x
)
print(f"Утверждение 1: ∀x ∃y (x + y = 0) на M={M}")
print(f"  Результат: {'T' if all_x_exists_y else 'F'}")
print(f"  (T, потому что для каждого x берём y = -x, и x + (-x) = 0)")

print()

# Утверждение 2: ∃y ∀x P(x,y) — существует y, что для всех x: x+y=0
# Проверяем, есть ли y, для которого P истинно при всех x
exists_y_all_x = any(
    all(P(x, y) for x in M)                    # ∀x: P(x, y)
    for y in M                                  # ∃y
)
print(f"Утверждение 2: ∃y ∀x (x + y = 0) на M={M}")
print(f"  Результат: {'T' if exists_y_all_x else 'F'}")
print(f"  (F, потому что не существует 'универсального' y, компенсирующего любой x)")

print()
print("=" * 60)
print("ВЫВОД: ∀x ∃y  ≠  ∃y ∀x")
print("Порядок кванторов ОПРЕДЕЛЯЕТ смысл утверждения!")
print()

# Предел последовательности: x_n = 1/n → 0
# Проверяем: ∀ε>0 ∃M: ∀n>M |1/n - 0| < ε
print("=" * 60)
print("Предел последовательности x_n = 1/n → 0")
print()

def x_n(n):
    """Член последовательности: 1/n."""
    return 1.0 / n

# Проверяем для нескольких ε
epsilons = [0.1, 0.01, 0.001, 0.0001]
print(f"{'ε':>10} | {'найдено M':>10} | {'проверка ∀n>M: |1/n| < ε':>30}")
print("-" * 60)

for eps in epsilons:
    # Ищем M такое, что для всех n > M: |1/n| < eps
    found_M = None
    for M_cand in range(1, 10000):
        if all(abs(x_n(n)) < eps for n in range(M_cand + 1, M_cand + 100)):
            found_M = M_cand
            break
    if found_M is not None:
        # Проверяем для нескольких n > M
        check_n = [found_M + 1, found_M + 10, found_M + 100]
        check_str = ', '.join(f"n={n}: {abs(x_n(n)):.6f}" for n in check_n)
        print(f"{eps:>10.4f} | {found_M:>10} | {check_str}")

print()
print("Профессор: для каждого ε находится своё M. Чем меньше ε, тем больше M.")
print("Это и есть смысл предела: ∀ε ∃M — M зависит от ε.")
````

## 14. Зачем вся эта логика в ML

Кажется, что логика высказываний — это про философов. Но на самом деле **вся ML-практика** пронизана логикой.

### 1. Фильтрация датасета

```python
df[(df.age > 18) & (df.income > 50000) & ~(df.city == 'Moscow')]
```

Это **логическая формула**: $(\text{age} > 18) \land (\text{income} > 50000) \land \neg(\text{city} = \text{Moscow})$.

### 2. Решающие деревья

Каждый узел дерева — это предикат $P(x)$: «$x_j > t$». Лист дерева определяется конъюнкцией предикатов на пути от корня. Любое решающее дерево — это DNF (дизъюнктивная нормальная форма) — ИЛИ нескольких И.

### 3. SQL-запросы

```sql
SELECT * FROM users
WHERE age > 18 AND (city = 'Moscow' OR city = 'SPb') AND NOT banned;
```

Это $P_1 \land (P_2 \lor P_3) \land \neg P_4$.

### 4. Логистическая регрессия

Решение «класс 1» принимается, если $w \cdot x + b > 0$. Это предикат $P(x) = (w \cdot x + b > 0)$.

### 5. Правила классификации

«Если возраст > 65 И есть хроническое заболевание, то высокий риск» — это импликация $P \land Q \to R$.

### 6. Контракты моделей

При валидации модели часто формулируют свойства в виде $\forall$-утверждений:
- «Для всех объектов из тестовой выборки accuracy ≥ 0.95»
- «Не существует группы пользователей, где fairness < 0.8»

Эти утверждения можно **проверять** через `all()` и `any()`.

### 7. Loss-функции и условия

MSE: $L = (y - \hat{y})^2$. Условие «loss велик» — это предикат. Условие «остановить обучение» — это импликация: «если loss не уменьшается 5 эпох подряд, то stop».

### 8. Мульти-классовая классификация

«Если класс 1, ИЛИ класс 2, И НЕ класс 3» — это логическая формула над предсказаниями.

### 9. Активации нейронов

ReLU: $f(x) = x$ если $x > 0$, иначе $0$. Это «если $x > 0$, то $f(x) = x$, иначе $f(x) = 0$» — импликация с двумя ветками.

### 10. Кванторы в определениях

- «Сходимость SGD»: $\forall \varepsilon \, \exists T$ — после $T$ шагов loss отличается от минимума менее чем на $\varepsilon$.
- «PAC-обучаемость»: $\forall \varepsilon, \delta \, \exists m$ — для любого $\varepsilon, \delta$ найдётся размер выборки $m$, такой что с вероятностью $1-\delta$ ошибка меньше $\varepsilon$.

> **Главный вывод:** логика — это **язык**, на котором формулируются все ML-модели, свойства, метрики. Если ты не владеешь этим языком — ты не сможешь ни доказать, ни опровергнуть утверждение о своей модели.

## 15. Сводная шпаргалка

### Операции и их таблицы

| Операция | Символ | Python | Истина когда... |
|---|---|---|---|
| Отрицание | $\neg P$ | `not p` | $P$ ложно |
| Конъюнкция | $P \land Q$ | `p and q` | оба истинны |
| Дизъюнкция | $P \lor Q$ | `p or q` | хотя бы один истинен |
| XOR | $P \oplus Q$ | `p != q` | значения различны |
| Импликация | $P \to Q$ | `(not p) or q` | не (P=И и Q=Л) |

### Кванторы

| Квантор | Символ | Python | Истина когда... |
|---|---|---|---|
| Всеобщности | $\forall x \, P(x)$ | `all(P(x) for x in M)` | $P$ истинно для всех |
| Существования | $\exists x \, P(x)$ | `any(P(x) for x in M)` | $P$ истинно хотя бы для одного |

### Главные законы

- **Де Морган 1:** $\neg(P \land Q) \equiv \neg P \lor \neg Q$
- **Де Морган 2:** $\neg(P \lor Q) \equiv \neg P \land \neg Q$
- **Двойное отрицание:** $\neg(\neg P) \equiv P$
- **Импликация:** $P \to Q \equiv \neg P \lor Q$
- **Отрицание импликации:** $\neg(P \to Q) \equiv P \land \neg Q$
- **Исключённое третье:** $P \lor \neg P \equiv$ И
- **Противоречие:** $P \land \neg P \equiv$ Л
- **Отрицание ∀:** $\neg(\forall x \, P) \equiv \exists x \, (\neg P)$
- **Отрицание ∃:** $\neg(\exists x \, P) \equiv \forall x \, (\neg P)$

### Три вопроса, чтобы выбрать операцию

1. **«Хочу, чтобы было и то, и другое одновременно»?** → конъюнкция $\land$
2. **«Хочу, чтобы хотя бы одно»?** → дизъюнкция $\lor$
3. **«Хочу, чтобы ровно одно (но не оба)»?** → XOR $\oplus$

### Четыре вопроса для кванторов

1. **«Все ли удовлетворяют»?** → $\forall$ (`all`)
2. **«Хотя бы один удовлетворяет»?** → $\exists$ (`any`)
3. **«Не все»?** → $\neg \forall \equiv \exists \neg$
4. **«Никто не»?** → $\neg \exists \equiv \forall \neg$

````python
# ============================================================
# Интерактивный тест — все вопросы и варианты плейнтекстом
# ============================================================
# Запусти ячейку (Shift+Enter), чтобы начать тест.

import ipywidgets as widgets
from IPython.display import display, clear_output

# Вопросы: (вопрос, [варианты], индекс правильного, пояснение)
# ВАЖНО: всё плейнтекстом с Unicode — НЕТ LaTeX, чтобы рендерилось везде
QUESTIONS = [
    {
        'q': 'Что такое ВЫСКАЗЫВАНИЕ?',
        'options': [
            'Любое предложение на любом языке',
            'Утверждение, о котором можно сказать истинно оно или ложно',
            'Вопрос, на который можно ответить да или нет',
            'Эмоциональное восклицание',
        ],
        'correct': 1,
        'expl': 'Высказывание — это повествовательное утверждение, к которому применимо «истинно» или «ложно». «Который час?» — не высказывание (это вопрос). «Сократ — философ» — высказывание.'
    },
    {
        'q': 'Чему равно ¬(¬P) (двойное отрицание)?',
        'options': ['Всегда И (истина)', 'Всегда Л (ложь)', 'Само P', 'Не определено'],
        'correct': 2,
        'expl': 'Двойное отрицание = утверждение. «Не не дождь» = «дождь». ¬(¬P) ≡ P.'
    },
    {
        'q': 'Дано: P = И (истина), Q = Л (ложь). Чему равно P ∧ Q (конъюнкция)?',
        'options': ['И', 'Л', 'Не определено', 'Зависит от контекста'],
        'correct': 1,
        'expl': 'Конъюнкция (И) истинна ТОЛЬКО когда оба истинны. Раз Q = Л, то P ∧ Q = Л.'
    },
    {
        'q': 'Дано: P = Л, Q = Л. Чему равно P → Q (импликация)?',
        'options': ['И (истина)', 'Л (ложь)', 'Не определено', 'Зависит от Q'],
        'correct': 0,
        'expl': 'Импликация ложна ТОЛЬКО когда P=И и Q=Л. Во всех остальных случаях (включая P=Л, Q=Л) — истинна. «Из лжи следует что угодно».'
    },
    {
        'q': 'В каком единственном случае импликация P → Q ложна?',
        'options': [
            'Когда P = И и Q = И',
            'Когда P = И и Q = Л',
            'Когда P = Л и Q = И',
            'Когда P = Л и Q = Л',
        ],
        'correct': 1,
        'expl': 'P → Q ложно только тогда, когда посылка P истинна, а заключение Q ложно. Это единственный случай «нарушенного обещания».'
    },
    {
        'q': 'Какой закон де Моргана верный?',
        'options': [
            '¬(P ∧ Q) ≡ ¬P ∨ ¬Q  (отрицание конъюнкции = дизъюнкция отрицаний)',
            '¬(P ∧ Q) ≡ P ∨ Q',
            '¬(P ∧ Q) ≡ ¬P ∧ ¬Q',
            '¬(P ∧ Q) ≡ P ∧ Q',
        ],
        'correct': 0,
        'expl': 'НЕ (P И Q) = (НЕ P) ИЛИ (НЕ Q). Пример: «НЕ (дождь И снег)» = «НЕ дождь ИЛИ НЕ снег» — достаточно, чтобы одного из них не было.'
    },
    {
        'q': 'Чему равно ¬(∀x P(x)) (отрицание квантора всеобщности)?',
        'options': [
            '∀x ¬P(x)  — для всех x: не P',
            '∃x ¬P(x)  — существует x, для которого не P',
            '∃x P(x)   — существует x, для которого P',
            '∀x P(x)   — без изменений',
        ],
        'correct': 1,
        'expl': '«НЕ (все сдали)» = «существует хотя бы один, кто НЕ сдал». Квантор меняется на парный, предикат отрицается: ¬(∀x P) ≡ ∃x ¬P.'
    },
    {
        'q': 'Дано: P = И, Q = И. Чему равно P ⊕ Q (XOR)?',
        'options': ['И (истина)', 'Л (ложь)', 'Не определено', 'Зависит от P'],
        'correct': 1,
        'expl': 'XOR истинен, когда P и Q РАЗНЫЕ. Если оба И (или оба Л) — XOR ложен. Здесь P=Q=И, значит ⊕ = Л.'
    },
    {
        'q': 'Что верно для кванторов ∀x ∃y P(x,y) и ∃y ∀x P(x,y)?',
        'options': [
            'Они всегда эквивалентны',
            'Они могут различаться: порядок кванторов важен',
            'Второе всегда сильнее первого',
            'Первое всегда сильнее второго',
        ],
        'correct': 1,
        'expl': 'Порядок кванторов важен! ∀x ∃y: «для каждого x есть свой y» (y зависит от x). ∃y ∀x: «есть один y для всех x» (y не зависит). Это разные утверждения.'
    },
    {
        'q': 'В Python what — аналог квантора существования ∃?',
        'options': ['all()', 'any()', 'filter()', 'map()'],
        'correct': 1,
        'expl': 'any(P(x) for x in M) возвращает True, если хотя бы для одного x предикат истинен. Это и есть ∃. Аналог ∀ — all().'
    },
]

# Состояние теста
state = {'index': 0, 'score': 0, 'answers': []}

# Виджеты
question_html = widgets.HTML()
options_radio = widgets.RadioButtons(options=[], layout=widgets.Layout(width='80%'))
feedback_html = widgets.HTML()
next_button = widgets.Button(description='Ответить →', button_style='primary',
                              layout=widgets.Layout(width='200px'))
progress_label = widgets.HTML()

def render_question():
    """Отрисовка текущего вопроса."""
    i = state['index']
    q = QUESTIONS[i]
    question_html.value = (
        f'<h3 style="color:#2c3e50">Вопрос {i+1} из {len(QUESTIONS)}</h3>'
        f'<p style="font-size:14px">{q["q"]}</p>'
    )
    options_radio.options = q['options']
    options_radio.value = None
    feedback_html.value = ''
    next_button.description = 'Ответить →'
    progress_label.value = f'<small>Счёт: {state["score"]}/{state["index"]}</small>'

def on_next(b):
    """Обработка нажатия на кнопку Ответить."""
    i = state['index']
    q = QUESTIONS[i]

    if options_radio.value is None:
        feedback_html.value = '<span style="color:#e74c3c">⚠️ Выбери вариант ответа!</span>'
        return

    chosen = q['options'].index(options_radio.value)
    correct = chosen == q['correct']

    if correct:
        state['score'] += 1
        feedback_html.value = (
            f'<div style="background:#eafaf1;padding:10px;border-left:4px solid #27ae60">'
            f'<b style="color:#27ae60">✓ Верно!</b><br>'
            f'<small>{q["expl"]}</small></div>'
        )
    else:
        correct_opt = q['options'][q['correct']]
        feedback_html.value = (
            f'<div style="background:#fdedee;padding:10px;border-left:4px solid #e74c3c">'
            f'<b style="color:#e74c3c">✗ Неверно.</b> Правильный ответ: <b>{correct_opt}</b><br>'
            f'<small>{q["expl"]}</small></div>'
        )

    state['answers'].append({
        'q': q['q'], 'chosen': chosen,
        'correct': q['correct'], 'ok': correct
    })

    if i + 1 < len(QUESTIONS):
        next_button.description = 'Следующий вопрос →'
        next_button.on_click(go_next, remove=True)
        next_button.on_click(go_next)
    else:
        next_button.description = 'Показать результат 🏁'
        next_button.on_click(show_results, remove=True)
        next_button.on_click(show_results)

def go_next(b):
    """Переход к следующему вопросу."""
    state['index'] += 1
    next_button.on_click(go_next, remove=True)
    next_button.on_click(on_next)
    render_question()

def show_results(b):
    """Финальный экран с результатом."""
    score = state['score']
    total = len(QUESTIONS)
    pct = score / total * 100

    if pct >= 90:
        grade, color, emoji = 'Отлично!', '#27ae60', '🎉'
    elif pct >= 70:
        grade, color, emoji = 'Хорошо', '#3498db', '👍'
    elif pct >= 50:
        grade, color, emoji = 'Удовлетворительно', '#f39c12', '🤔'
    else:
        grade, color, emoji = 'Нужно перечитать', '#e74c3c', '📚'

    html = f'''
    <div style="background:{color};color:white;padding:20px;border-radius:10px;text-align:center">
        <h2>{emoji} {grade}</h2>
        <p style="font-size:18px">Твой результат: <b>{score} / {total}</b> ({pct:.0f}%)</p>
    </div>
    <h4 style="margin-top:20px">Разбор ответов:</h4>
    <ol>
    '''
    for ans in state['answers']:
        icon = '✓' if ans['ok'] else '✗'
        clr = '#27ae60' if ans['ok'] else '#e74c3c'
        html += f'<li style="color:{clr}">{icon} {ans["q"][:80]}...</li>'
    html += '</ol>'

    clear_output()
    display(widgets.HTML(html))

# Стартуем тест
next_button.on_click(on_next)
render_question()

display(widgets.VBox([
    progress_label,
    question_html,
    options_radio,
    feedback_html,
    next_button,
]))
````
