# Тема 1. Что такое высказывание

## 1.1. Постановка проблемы и практический пример

В разговоре мы говорим разные вещи: вопросы, восклицания, утверждения. Математика интересуется только **утверждениями, о которых можно сказать «истинно» или «ложно»**. Не «верю/не верю», не «нравится/не нравится», а строго бинарно: True или False.

Например, «2 + 2 = 4» — высказывание (истинно). «Идёт ли дождь?» — не высказывание (это вопрос). «Какая красота!» — не высказывание (восклицание). «На Марсе есть жизнь» — высказывание (истинность неизвестна, но в принципе определима).

## 1.3. Интуитивное объяснение и аналогия

> **Аналогия:** Высказывание — это **лампочка**: либо горит (True), либо нет (False). Промежуточных состояний нет. Если про фразу нельзя сказать «горит/не горит» — это не высказывание (как диммер с плавной регулировкой — не лампочка в нашем смысле).

*Юмор от Профессора:* «Жена спрашивает: "Ты меня любишь?" — это не высказывание, а ловушка. Какой ответ ни дай, она либо истинна, либо ложна — но не для тебя.»

## 1.5. Техническое объяснение

**Высказывание** — это утверждение, для которого имеет смысл говорить об истинности или ложности.

Обозначения:
- $p, q, r, \ldots$ — переменные для высказываний (пропозициональные переменные).
- $\top$ (или T, 1) — истина (True).
- $\bot$ (или F, 0) — ложь (False).

**Не являются высказываниями:**
- Вопросы: «Который час?»
- Восклицания: «Ого!»
- Приказы: «Иди сюда!»
- Парадоксы: «Это утверждение ложно» (если истинно — то ложно; если ложно — то истинно).

## 1.6. Внутреннее устройство

**Закон исключённого третьего.** Для любого высказывания $p$ верно: либо $p$, либо $\neg p$ (не $p$). Третьего не дано.

**Закон противоречия.** Невозможно, чтобы $p$ и $\neg p$ были одновременно истинны.

**Двоичная логика.** Мы работаем в **классической двузначной логике**: только True и False. Существуют и другие (многозначные, нечёткие), но они — за рамками этой лекции.

## 1.7. Визуальная схема

```
Высказывание = лампочка (горит / не горит)

         True (1, T, ⊤)        False (0, F, ⊥)
            ┌────┐                 ┌────┐
            │ 💡 │                 │ 💡 │
            └────┘                 └────┘
           горит                 не горит

"2+2=4" → True          "2+2=5" → False

НЕ высказывания:
  - "Который час?" (вопрос)
  - "Иди сюда!" (приказ)
  - "Ого!" (восклицание)
  - "Это ложно" (парадокс)
```

## 1.8. Рабочий пример

````python
# ============================================================
# Высказывания в Python
# ============================================================

# Примеры высказываний (булевых выражений)
p = (2 + 2 == 4)                    # True
q = (2 + 2 == 5)                    # False
r = ("москва" == "москва")          # True

print(f"p = (2+2==4) = {p}")
print(f"q = (2+2==5) = {q}")
print(f"r = ('москва'=='москва') = {r}")

# Тип bool в Python — это именно True/False
print(f"\nТип p: {type(p).__name__}")
print(f"True это int? {isinstance(True, int)} (True наследует от int!)")
print(f"True + True = {True + True}  (в Python bool — подкласс int)")

# НЕ высказывание (вопрос — не True/False)
question = "Который час?"
print(f"\n'{question}' — не высказывание (нет True/False)")

# Парадокс: "Это утверждение ложно"
# Если оно истинно → оно ложно → противоречие
# Если оно ложно → оно истинно → противоречие
print("\nПарадокс лжеца: 'Это утверждение ложно' — не высказывание в классической логике.")
````

# Тема 2. Простые и составные высказывания

## 2.1. Постановка проблемы и практический пример

«Идёт дождь» — простое высказывание. «Идёт дождь И дует ветер» — составное: оно состоит из двух простых, соединённых союзом «И». Истинность составного зависит от истинности частей.

В логике есть **5 основных операций** для соединения простых высказываний в составные: $\neg$ (НЕ), $\land$ (И), $\lor$ (ИЛИ), $\oplus$ (XOR), $\to$ (импликация). Каждой — посвящена отдельная тема.

## 2.3. Интуитивное объяснение и аналогия

> **Аналогия:** Простые высказывания — как **атомы**, составные — как **молекулы**. Молекула воды $H_2O$ состоит из 2 атомов водорода и 1 атома кислорода, соединённых связями. Точно так же составное высказывание «$p \land q$» состоит из 2 простых, соединённых операцией $\land$.

*Юмор от Профессора:* «Программист на вопрос "ты голоден?" отвечает "True". На вопрос "хочешь пиццу или бургер?" — "True OR True", то есть "True". Жена программиста уже привыкла.»

## 2.5. Техническое объяснение

**Простое высказывание** (атом, пропозициональная переменная) — не разбивается на части. Обозначается $p, q, r, \ldots$

**Составное высказывание** (формула) — строится из простых с помощью операций:
- $\neg p$ — отрицание (НЕ $p$)
- $p \land q$ — конъюнкция (И)
- $p \lor q$ — дизъюнкция (ИЛИ)
- $p \oplus q$ — строгая дизъюнкция (XOR, либо... либо...)
- $p \to q$ — импликация (если $p$, то $q$)
- $p \leftrightarrow q$ — эквивалентность ($p$ тогда и только тогда, когда $q$)

**Скобки.** Как в арифметике, скобки задают порядок:
- $(p \land q) \lor r$ — сначала $\land$, потом $\lor$.
- $p \land (q \lor r)$ — сначала $\lor$, потом $\land$.
Это **разные** высказывания!

**Приоритет операций** (по убыванию):
$$\neg > \land > \lor > \to > \leftrightarrow$$

## 2.6. Внутреннее устройство

**Истинность составного** — функция от истинности частей. Например, $p \land q$ истинно **только** когда оба $p$ и $q$ истинны.

Таблица истинности — основной инструмент. Для формулы с $n$ переменными таблица имеет $2^n$ строк (все комбинации True/False).

**Пример.** Для $(p \land q) \lor \neg r$ (3 переменные) — $2^3 = 8$ строк.

## 2.7. Визуальная схема

```
Простые высказывания (атомы):     p, q, r, ...
   ↓
Соединяем операциями:             ¬p, p ∧ q, p ∨ q, p ⊕ q, p → q, p ↔ q
   ↓
Составные (молекулы):             (p ∧ q) ∨ ¬r,  ¬(p ∨ q) → (r ∧ s),  ...

Приоритет: ¬ > ∧ > ∨ > → > ↔

Пример: p ∨ q ∧ r  без скобок = p ∨ (q ∧ r)  (∝ выше чем ∨)
        p → q ∨ r  без скобок = p → (q ∨ r)  (∨ выше чем →)

Таблица истинности для n переменных: 2^n строк
```

## 2.8. Рабочий пример

````python
# ============================================================
# Простые и составные высказывания
# ============================================================

# Простые
p = True                            # "Идёт дождь"
q = False                           # "Дует ветер"
r = True                            # "Холодно"

# Составные
not_p = not p                       # ¬p: "Дождя нет"
p_and_q = p and q                   # p ∧ q: "Дождь И ветер"
p_or_q = p or q                     # p ∨ q: "Дождь ИЛИ ветер"
p_xor_q = (p != q)                  # p ⊕ q: "Либо дождь, либо ветер (но не оба)"
p_implies_q = (not p) or q          # p → q: "Если дождь, то ветер"  (об этом — тема 7)

print(f"p = {p}, q = {q}, r = {r}")
print(f"¬p       = {not_p}")
print(f"p ∧ q    = {p_and_q}")
print(f"p ∨ q    = {p_or_q}")
print(f"p ⊕ q    = {p_xor_q}")
print(f"p → q    = {p_implies_q}")

# Составное из 3 простых
formula = (p and q) or (not r)      # (p ∧ q) ∨ ¬r
print(f"\n(p ∧ q) ∨ ¬r = {formula}")

# Приоритеты: ¬ > ∧ > ∨ > → > ↔
# p ∨ q ∧ r без скобок = p ∨ (q ∧ r)
a, b, c = True, False, True
no_brackets = a or (b and c)        # Python сам применяет правильный приоритет
explicit = a or (b and c)
print(f"\np ∨ q ∧ r без скобок = {no_brackets}")
print(f"p ∨ (q ∧ r) явно     = {explicit} (совпадает)")
print(f"(p ∨ q) ∧ r          = {(a or b) and c} (другое значение!)")
````

## 2.9. Практический эксперимент

Проверим все 8 комбинаций True/False для 3 переменных:

````python
# ============================================================
# Эксперимент: все 8 комбинаций 3 переменных
# ============================================================
import itertools

# Все комбинации (p, q, r) — 2^3 = 8 штук
combos = list(itertools.product([True, False], repeat=3))

print(f"{'p':>6} {'q':>6} {'r':>6} | {'¬p':>6} {'p∧q':>6} {'p∨q':>6} {'(p∧q)∨¬r':>10}")
print("-" * 60)
for p, q, r in combos:
    not_p = not p
    p_and_q = p and q
    p_or_q = p or q
    formula = (p and q) or (not r)
    print(f"{str(p):>6} {str(q):>6} {str(r):>6} | {str(not_p):>6} {str(p_and_q):>6} {str(p_or_q):>6} {str(formula):>10}")

# Профессор: "Для n переменных таблица имеет 2^n строк. Для 5 переменных — уже 32."
#             "Это быстро растёт, поэтому формулы с >5 переменными редко выписывают целиком."
````

# Тема 3. Отрицание $\neg$

## 3.1. Постановка проблемы и практический пример

«Идёт дождь» — это $p$. А «дождя нет» — это $\neg p$ (читается «не пэ»). Если $p$ истинно, то $\neg p$ ложно, и наоборот. Отрицание **переворачивает** истинность.

В Python: `not p`. В SQL: `NOT p`. В C/Java: `!p`. Везде одно и то же — операция унарная (один аргумент), меняет True на False и наоборот.

## 3.3. Интуитивное объяснение и аналогия

> **Аналогия:** Отрицание — это **выключатель**: что горело — гаснет, что не горело — загорается. Одно нажатие — одно переключение. Двойное нажатие — возврат к исходному состоянию ($\neg \neg p = p$).

*Юмор от Профессора:* «Программист, которому сказали "не не делай этого", запутался. Логик — понял: двойное отрицание = утверждение. Делай.»

## 3.5. Техническое объяснение

**Отрицание** $\neg p$ (читается «не $p$»):
| $p$ | $\neg p$ |
|---|---|
| T | F |
| F | T |

**Свойства:**
- **Двойное отрицание:** $\neg \neg p \equiv p$ (закон инволюции).
- **Закон исключённого третьего:** $p \lor \neg p \equiv \top$ (всегда истина).
- **Закон противоречия:** $p \land \neg p \equiv \bot$ (всегда ложь).

## 3.6. Внутреннее устройство

**Почему $\neg \neg p = p$?** Если перевернуть True/False дважды — вернёшься к исходному. Это как дважды отразить в зеркале.

**В Python:** `not (not p) == p` — всегда верно.

**В естественном языке:** двойное отрицание **не всегда** равно утверждению. «Я не могу не согласиться» = «я согласен» (равно). Но «он не некрасивый» ≠ «он красивый» (может быть, «обычный»). В логике такого нет: $\neg \neg p = p$ строго.

## 3.7. Визуальная схема

```
Отрицание ¬p — это "переключатель":

        p        ¬p
        T   →    F     (True → False)
        F   →    T     (False → True)

Двойное отрицание:
        p   ¬p   ¬¬p
        T    F    T     ← вернулись к p
        F    T    F     ← вернулись к p

Закон исключённого третьего:
        p ∨ ¬p = T всегда
        T ∨ F = T
        F ∨ T = T

Закон противоречия:
        p ∧ ¬p = F всегда
        T ∧ F = F
        F ∧ T = F
```

## 3.8. Рабочий пример

````python
# ============================================================
# Отрицание в Python
# ============================================================

def neg(p):
    # Отрицание: НЕ p. Возвращает True, если p=False, и наоборот.
    return not p

# Таблица истинности
print(f"{'p':>6} | {'¬p':>6}")
print("-" * 16)
for p in [True, False]:
    print(f"{str(p):>6} | {str(neg(p)):>6}")

# Двойное отрицание
print(f"\nДвойное отрицание:")
print(f"{'p':>6} | {'¬p':>6} | {'¬¬p':>6}")
print("-" * 26)
for p in [True, False]:
    print(f"{str(p):>6} | {str(neg(p)):>6} | {str(neg(neg(p))):>6}")

# Проверка свойств
print("\nЗакон инволюции: ¬¬p ≡ p?")
for p in [True, False]:
    print(f"  p={p}: ¬¬p={neg(neg(p))}, совпадает? {neg(neg(p)) == p}")

print("\nЗакон исключённого третьего: p ∨ ¬p ≡ T?")
for p in [True, False]:
    print(f"  p={p}: p ∨ ¬p = {p or neg(p)} (всегда True)")

print("\nЗакон противоречия: p ∧ ¬p ≡ F?")
for p in [True, False]:
    print(f"  p={p}: p ∧ ¬p = {p and neg(p)} (всегда False)")
````

## 3.9. Практический эксперимент

Реальные примеры:

````python
# ============================================================
# Эксперимент: отрицание в реальных высказываниях
# ============================================================

# Реальные высказывания
statements = [
    ("Идёт дождь", True),
    ("Я программист", False),
    ("Москва — столица России", True),
    ("2 + 2 = 5", False),
]

print(f"{'Высказывание':<30} | {'p':>6} | {'¬p':>6}")
print("-" * 55)
for text, val in statements:
    print(f"{text:<30} | {str(val):>6} | {str(not val):>6}")

# Применение в фильтрации данных
print("\n--- Фильтрация данных через ¬ ---")
import numpy as np
data = np.array([3, -1, 0, 5, -2, 7, -4])

# Положительные числа: x > 0
positives = data[data > 0]
# НЕ положительные (¬(x > 0)): x ≤ 0
not_positives = data[not (data > 0)] if False else data[~(data > 0)]  # используем ~ для numpy
print(f"Данные: {data}")
print(f"Положительные (x > 0): {positives}")
print(f"НЕ положительные (¬(x > 0) = x ≤ 0): {not_positives}")

# Профессор: "В numpy оператор ~ это побитовое НЕ, для булевых массивов работает как ¬."
````

## 3.11. Типичные ошибки

## 3.12. Краткий вывод и переход

- $\neg \neg p = p$.
- $p \lor \neg p = \top$, $p \land \neg p = \bot$.
- В Python: `not p`.

# Тема 4. Конъюнкция $\land$

## 4.1. Постановка проблемы и практический пример

«Идёт дождь И дует ветер». Когда это истинно? Только если **оба** условия выполнены. Если дождя нет — ложно. Если ветра нет — ложно. Если нет ни того, ни другого — тем более ложно.

Это **конъюнкция** — логическое «И». Обозначается $\land$.

## 4.3. Интуитивное объяснение и аналогия

> **Аналогия:** Конъюнкция — это **пересечение списков**. У тебя список друзей, у брата — свой. «Друзья И твои, И брата» = пересечение списков. В пересечении только те, кто есть в обоих.

*Юмор от Профессора:* «Конъюнкция — строгий экзаменатор. Чтобы сдать, нужно сдать ВСЕ предметы. Один долг — и ты отчислен.»

## 4.5. Техническое объяснение

**Конъюнкция** $p \land q$ (читается «$p$ и $q$»):

| $p$ | $q$ | $p \land q$ |
|---|---|---|
| T | T | **T** |
| T | F | F |
| F | T | F |
| F | F | F |

**Истинна только когда оба истинны.**

**Свойства:**
- **Коммутативность:** $p \land q \equiv q \land p$.
- **Ассоциативность:** $(p \land q) \land r \equiv p \land (q \land r)$.
- **Идемпотентность:** $p \land p \equiv p$.
- **Тождество:** $p \land \top \equiv p$.
- **Поглощение (доминирование):** $p \land \bot \equiv \bot$.
- **Поглощение (сокращение):** $p \land (p \lor q) \equiv p$.

## 4.6. Внутреннее устройство

**В Python:** `p and q`. Важная особенность: `and` — **ленивый**. Если `p = False`, Python не вычисляет `q` (всё равно False).

**Применения:**
- Фильтры: `df[(df.x > 0) & (df.y < 10)]` — конъюнкция условий.
- Пароли: `if len(p) >= 8 and has_digit(p) and has_upper(p)`.
- Логика поиска: `WHERE title LIKE '%Python%' AND year > 2020`.

**Частный случай — нескольких переменных:** $p_1 \land p_2 \land \ldots \land p_n$ истинна, только если **все** $p_i$ истинны.

## 4.7. Визуальная схема

```
Конъюнкция p ∧ q — "И" (строгий экзаменатор):

p    q    p∧q
T    T     T   ← только оба истинны
T    F     F
F    T     F
F    F     F

Круги Эйлера:
   ┌──────────┐
  / A \      / B  /     \    /      \     \  /     /
  \     \/     /
   \    /\    /
    \  /  \  /
     \/    \/
     ←←←→→→
     A∩B
(пересечение = только то, что в обоих)

Свойства:
- p ∧ q = q ∧ p           (коммутативность)
- p ∧ p = p               (идемпотентность)
- p ∧ T = p               (тождество)
- p ∧ F = F               (доминирование)
```

## 4.8. Рабочий пример

````python
# ============================================================
# Конъюнкция в Python
# ============================================================

def conj(p, q):
    # Конъюнкция: p ∧ q. Истинна только если оба истинны.
    return p and q

# Таблица истинности
print(f"{'p':>6} {'q':>6} | {'p∧q':>6}")
print("-" * 24)
for p in [True, False]:
    for q in [True, False]:
        print(f"{str(p):>6} {str(q):>6} | {str(conj(p, q)):>6}")

# Ленивость: and не вычисляет второй аргумент, если первый False
def side_effect():
    print("    (вызвана side_effect)")
    return True

print("\n--- Ленивость 'and' ---")
print("False and side_effect():")
result = False and side_effect()       # side_effect НЕ вызывается
print(f"Результат: {result}")

print("\nTrue and side_effect():")
result = True and side_effect()        # side_effect вызывается
print(f"Результат: {result}")

# Профессор: "Ленивость спасает от ошибок: 'if x is not None and x.field > 0' не упадёт."
````

## 4.9. Практический эксперимент

Применение конъюнкции в фильтрации данных:

````python
# ============================================================
# Эксперимент: фильтрация через конъюнкцию
# ============================================================

# Список студентов: (имя, оценка, посещаемость)
students = [
    ("Анна",   85, 95),
    ("Боря",   72, 80),
    ("Вася",   90, 60),
    ("Галя",   65, 90),
    ("Дима",   88, 85),
    ("Ева",    55, 70),
    ("Женя",   95, 100),
]

# Конъюнкция условий: оценка ≥ 75 И посещаемость ≥ 80
print("Студенты с оценкой ≥ 75 И посещаемостью ≥ 80:")
print(f"{'Имя':<10} {'Оценка':>8} {'Посещ.':>8} | {'≥75 ∧ ≥80':>10}")
print("-" * 45)
for name, grade, attendance in students:
    cond1 = grade >= 75
    cond2 = attendance >= 80
    both = cond1 and cond2                     # конъюнкция
    marker = "✓" if both else "✗"
    print(f"{name:<10} {grade:>8} {attendance:>8} | {marker:>10}")

# Количество прошедших фильтр
passed = sum(1 for _, g, a in students if g >= 75 and a >= 80)
print(f"\nПрошло фильтр: {passed} из {len(students)}")

# Профессор: "Конъюнкция = AND в SQL, & в pandas, and в Python — везде одно и то же."
````

## 4.11. Типичные ошибки

- **Причина:** путают с дизъюнкцией.

## 4.12. Краткий вывод и переход

- $p \land q$ — конъюнкция, истинна только если оба истинны.
- Коммутативна, ассоциативна, идемпотентна.
- $p \land \top = p$, $p \land \bot = \bot$.
- В Python: `and` (ленивый).

# Тема 5. Дизъюнкция $\lor$

## 5.1. Постановка проблемы и практический пример

«Хочу чай ИЛИ кофе». Это истинно, если есть чай, ИЛИ есть кофе, ИЛИ есть оба. Ложно только если нет ни того, ни другого.

Это **дизъюнкция** — **нестрогое** логическое «ИЛИ». Обозначается $\lor$.

## 5.3. Интуитивное объяснение и аналогия

> **Аналогия:** Дизъюнкция — **добрый экзаменатор**. Чтобы сдать, нужно сдать хотя бы один предмет. Сдал один — молодец. Сдал все — отлично. Не сдал ни одного — отчислен.

*Юмор от Профессора:* «Дизъюнкция — это как бар "или пиво, или водка". Можно одно, можно другое, можно оба. Бармен не возражает. Только не пей за рулём.»

## 5.5. Техническое объяснение

**Дизъюнкция** $p \lor q$ (читается «$p$ или $q$»):

| $p$ | $q$ | $p \lor q$ |
|---|---|---|
| T | T | **T** |
| T | F | **T** |
| F | T | **T** |
| F | F | F |

**Истинна если хотя бы одно истинно** (или оба).

**Свойства:**
- **Коммутативность:** $p \lor q \equiv q \lor p$.
- **Ассоциативность:** $(p \lor q) \lor r \equiv p \lor (q \lor r)$.
- **Идемпотентность:** $p \lor p \equiv p$.
- **Тождество:** $p \lor \bot \equiv p$.
- **Доминирование:** $p \lor \top \equiv \top$.
- **Дистрибутивность:** $p \lor (q \land r) \equiv (p \lor q) \land (p \lor r)$.

## 5.6. Внутреннее устройство

**В Python:** `p or q` — **ленивый**. Если `p = True`, не вычисляет `q`.

**Различие строгой и нестрогой дизъюнкции:**
- Нестрогая ($\lor$): «ИЛИ» — допускает оба (чай ИЛИ кофе, можно оба).
- Строгая ($\oplus$): «ЛИБО... ЛИБО...» — исключает оба (выйти через дверь ИЛИ окно, нельзя оба).

**В естественном языке:** иногда «или» = строгое (меню в ресторане: первое ИЛИ второе — обычно не оба), иногда нестрогое (деньги ИЛИ карта — можно и то, и другое). В логике «$\lor$» — **всегда** нестрогое.

## 5.7. Визуальная схема

```
Дизъюнкция p ∨ q — "ИЛИ" (нестрогое, добрый экзаменатор):

p    q    p∨q
T    T     T   ← оба истинны — тоже True
T    F     T
F    T     T
F    F     F   ← только если оба ложны

Круги Эйлера (объединение):
   ┌──────────┐
  / A \\\\  / B  /     \\\\/      \     /////\     /
  \     /////     /
   \    /////    /
    \  /////    /
     \/////    /
     ←←←→→→
     A∪B (всё, что в A или B, включая пересечение)

Свойства:
- p ∨ q = q ∨ p           (коммутативность)
- p ∨ p = p               (идемпотентность)
- p ∨ F = p               (тождество)
- p ∨ T = T               (доминирование)
- p ∨ (q ∧ r) = (p ∨ q) ∧ (p ∨ r)  (дистрибутивность)
```

## 5.8. Рабочий пример

````python
# ============================================================
# Дизъюнкция в Python
# ============================================================

def disj(p, q):
    # Дизъюнкция: p ∨ q. Истинна если хотя бы одно истинно.
    return p or q

# Таблица истинности
print(f"{'p':>6} {'q':>6} | {'p∨q':>6}")
print("-" * 24)
for p in [True, False]:
    for q in [True, False]:
        print(f"{str(p):>6} {str(q):>6} | {str(disj(p, q)):>6}")

# Ленивость: or не вычисляет второй аргумент, если первый True
def side_effect():
    print("    (вызвана side_effect)")
    return False

print("\n--- Ленивость 'or' ---")
print("True or side_effect():")
result = True or side_effect()            # side_effect НЕ вызывается
print(f"Результат: {result}")

print("\nFalse or side_effect():")
result = False or side_effect()           # side_effect вызывается
print(f"Результат: {result}")

# Профессор: "Ленивость часто используется для значений по умолчанию:"
#             "name = user_input or 'Аноним'  — если user_input пусто, берём 'Аноним'."
````

## 5.9. Практический эксперимент

Дистрибутивность: $p \lor (q \land r) = (p \lor q) \land (p \lor r)$

````python
# ============================================================
# Эксперимент: проверка дистрибутивности
# ============================================================
import itertools

print("Проверка дистрибутивности: p ∨ (q ∧ r) ≡ (p ∨ q) ∧ (p ∨ r)")
print()
print(f"{'p':>6} {'q':>6} {'r':>6} | {'q∧r':>5} {'p∨(q∧r)':>9} | {'p∨q':>5} {'p∨r':>5} {'(p∨q)∧(p∨r)':>12} | совпадает?")
print("-" * 75)

all_match = True
for p, q, r in itertools.product([True, False], repeat=3):
    left = p or (q and r)
    right = (p or q) and (p or r)
    match = left == right
    if not match:
        all_match = False
    print(f"{str(p):>6} {str(q):>6} {str(r):>6} | {str(q and r):>5} {str(left):>9} | {str(p or q):>5} {str(p or r):>5} {str(right):>12} | {'✓' if match else '✗'}")

print(f"\nВсе 8 строк совпадают? {all_match}")

# Профессор: "Дистрибутивность ∨ относительно ∧ — это как распределительный закон в арифметике:"
#             "a × (b + c) = a×b + a×c. Только в логике '×' это ∧, а '+' это ∨."
````

## 5.10. Мини-проверка

**Вопрос 1.** Чему равно $T \lor F$?

<details><summary><b>Ответ</b></summary>
$T$. Дизъюнкция истинна, если хотя бы одно истинно.
</details>

**Вопрос 2.** Чему равно $p \lor \top$?

<details><summary><b>Ответ</b></summary>
$\top$. Если один из операндов всегда истинен, дизъюнкция — всегда истинна.
</details>

**Вопрос 3.** В Python: `True or some_function()` — вызовется ли `some_function()`?

<details><summary><b>Ответ</b></summary>
**Нет**. `or` ленивый: если первый операнд True, второй не вычисляется.
</details>

**Вопрос 4.** Чем $\lor$ отличается от $\oplus$?

<details><summary><b>Ответ</b></summary>
$\lor$ — нестрогое ИЛИ (допускает оба). $\oplus$ — строгое (исключает оба).
</details>

## 5.11. Типичные ошибки

**Ошибка 1.** «$p \lor q$ ложно, если оба истинны.»
- **Причина:** путают со строгим XOR.
- **Правильно:** $\lor$ — нестрогое, оба истинны тоже даёт $T$.

**Ошибка 2.** «В Python `|` и `or` — одно и то же.»
- **Причина:** путают побитовое и логическое.
- **Правильно:** `or` — логическое (ленивое, возвращает bool или первый truthy), `|` — побитовое (для int) или элементное (для numpy).

**Ошибка 3.** «Дизъюнкция — всегда строгое "или".»
- **Причина:** переносят разговорный смысл.
- **Правильно:** в логике $\lor$ — **нестрогое**. Строгое — это $\oplus$.

## 5.12. Краткий вывод и переход

- $p \lor q$ — дизъюнкция, истинна если **хотя бы одно** истинно.
- Коммутативна, ассоциативна, идемпотентна.
- $p \lor \bot = p$, $p \lor \top = \top$.
- Дистрибутивна относительно $\land$.
- В Python: `or` (ленивый).

**Чек-лист:**
- [ ] Я знаю таблицу истинности $\lor$.
- [ ] Я отличаю $\lor$ от $\oplus$.
- [ ] Я знаю свойства $\lor$.

**Переход:** Теперь разберём **строгое** «или» — XOR, обозначается $\oplus$.

# Тема 6. XOR $\oplus$

## 6.1. Постановка проблемы и практический пример

«Ты либо со мной, либо против меня» — исключающая альтернатива. Нельзя быть и тем, и другим. Нельзя быть ни тем, ни другим. Только одно из двух.

Это **строгое ИЛИ** — **XOR** (от англ. eXclusive OR). Обозначается $\oplus$.

## 6.3. Интуитивное объяснение и аналогия

> **Аналогия:** XOR — это **выбор блюда в ресторане**: первое ИЛИ второе, но не оба (за оба — двойная цена, не пройдёт). Либо рыба, либо мясо. Третьего (оба) не дано.

*Юмор от Профессора:* «XOR — это операция "или пан, или пропал". Если выбрал "и пан, и пропал" — ты и пан, и пропал, что невозможно по определению XOR.»

## 6.5. Техническое объяснение

**XOR** $p \oplus q$ (строгая дизъюнкция, читается «либо $p$, либо $q$»):

| $p$ | $q$ | $p \oplus q$ |
|---|---|---|
| T | T | F |
| T | F | **T** |
| F | T | **T** |
| F | F | F |

**Истинна если ровно одно из $p, q$ истинно.**

**Свойства:**
- **Коммутативность:** $p \oplus q \equiv q \oplus p$.
- **Ассоциативность:** $(p \oplus q) \oplus r \equiv p \oplus (q \oplus r)$.
- **Тождество:** $p \oplus \bot \equiv p$ (XOR с ложью = исходное).
- **Аннулирование:** $p \oplus \top \equiv \neg p$ (XOR с истиной = отрицание).
- **Самоаннулирование:** $p \oplus p \equiv \bot$ (XOR с собой = ложь).
- **Связь с $\neg, \lor, \land$:** $p \oplus q \equiv (p \lor q) \land \neg(p \land q)$.

## 6.6. Внутреннее устройство

**В Python:** `p ^ q` (или `p != q` для булевых значений).

**В электронике:** XOR — базовый элемент, используется в сумматорах. $a \oplus b$ = сумма без переноса: $0 \oplus 0 = 0$, $1 \oplus 0 = 1$, $0 \oplus 1 = 1$, $1 \oplus 1 = 0$ (а перенос — это $a \land b$).

**В криптографии:** XOR — основа многих шифров. Почему? Потому что $\text{message} \oplus \text{key} = \text{ciphertext}$, и $\text{ciphertext} \oplus \text{key} = \text{message}$ (восстановление). XOR обратим сам по себе.

## 6.7. Визуальная схема

```
XOR p ⊕ q — "либо... либо..." (строгое ИЛИ):

p    q    p⊕q
T    T     F   ← оба истинны — ТОЖЕ ЛОЖЬ (исключаем)
T    F     T   ← только одно истинно
F    T     T   ← только одно истинно
F    F     F

Круги Эйлера (симметрическая разность):
   ┌──────────┐
  / A \\\  / B  /     \\\//      \     ////\     /     только "полумесяцы"
  \   ////  \\   /     (без пересечения!)
   \ ////    \\ /
    ///        \\
     ←←         →→→
     AΔB = (A∖B) ∪ (B∖A)

Свойства:
- p ⊕ q = q ⊕ p           (коммутативность)
- p ⊕ p = F               (самоаннулирование)
- p ⊕ F = p               (тождество)
- p ⊕ T = ¬p              (XOR с T = отрицание)
```

## 6.8. Рабочий пример

````python
# ============================================================
# XOR в Python
# ============================================================

def xor(p, q):
    # XOR: p ⊕ q. Истинна если ровно одно из p, q истинно.
    return p != q                    # для булевых значений это и есть XOR

# Таблица истинности
print(f"{'p':>6} {'q':>6} | {'p⊕q':>6} | {'(p∨q)∧¬(p∧q)':>14}")
print("-" * 50)
for p in [True, False]:
    for q in [True, False]:
        direct = xor(p, q)
        via_basic = (p or q) and not (p and q)         # эквивалент через ∨, ∧, ¬
        print(f"{str(p):>6} {str(q):>6} | {str(direct):>6} | {str(via_basic):>14}")

# Проверка свойств
print("\n--- Свойства XOR ---")
print(f"p ⊕ p = F?  T⊕T = {xor(True, True)}, F⊕F = {xor(False, False)}  (всегда False)")
print(f"p ⊕ F = p?  T⊕F = {xor(True, False)}, F⊕F = {xor(False, False)}  (равно p)")
print(f"p ⊕ T = ¬p? T⊕T = {xor(True, True)}, F⊕T = {xor(False, True)}  (равно ¬p)")

# Применение: криптография (простой шифр)
print("\n--- Простой шифр на XOR ---")
message = [1, 0, 1, 1, 0, 1, 0, 0]      # 8 бит
key     = [1, 1, 0, 1, 0, 0, 1, 1]      # 8 бит ключа
cipher  = [m ^ k for m, k in zip(message, key)]   # шифруем
decoded = [c ^ k for c, k in zip(cipher, key)]    # расшифровываем

print(f"Сообщение:  {message}")
print(f"Ключ:       {key}")
print(f"Шифр:       {cipher}")
print(f"Расшифровка:{decoded}")
print(f"Совпадает с оригиналом? {decoded == message}")
````

## 6.9. Практический эксперимент

Многократный XOR:

````python
# ============================================================
# Эксперимент: многократный XOR
# ============================================================

# XOR нескольких переменных: истинна если НЕЧЁТНОЕ число True
import itertools

def xor_all(values):
    result = False
    for v in values:
        result = result != v       # XOR накапливаем
    return result

print("XOR нескольких переменных = True если НЕЧЁТНОЕ число True:")
print()
print(f"{'p':>6} {'q':>6} {'r':>6} | {'# True':>8} | {'p⊕q⊕r':>8}")
print("-" * 42)
for p, q, r in itertools.product([True, False], repeat=3):
    n_true = sum([p, q, r])
    xor_result = xor_all([p, q, r])
    print(f"{str(p):>6} {str(q):>6} {str(r):>6} | {n_true:>8} | {str(xor_result):>8}")

# Профессор: "XOR 3 переменных = True если нечётное число True. Это как 'бит чётности'."
#             "В CRC-кодах и контрольных суммах это используется повсеместно."

# Применение: проверка чётности
data = [1, 0, 1, 0, 1, 1, 0, 1]
parity = xor_all([bool(b) for b in data])
print(f"\nДанные: {data}")
print(f"XOR всех битов = {parity} (это бит чётности)")
print(f"Если передать данные + бит чётности, приёмник может проверить, не было ли ошибки.")
````

# Тема 7. Импликация $\to$

## 7.1. Постановка проблемы и практический пример

«Если пойдёт дождь, то я возьму зонт». Это **обещание**. Когда оно нарушено? Только если дождь пошёл, а зонт ты **не** взял. В остальных случаях — обещание не нарушено.

Это **импликация** — логическое «если..., то...». Обозначается $\to$. Это **самая контринтуитивная** операция, потому что в естественном языке мы часто вкладываем в «если... то...» дополнительный смысл (причинность), которого в логике нет.

## 7.3. Интуитивное объяснение и аналогия

> **Аналогия:** Импликация $p \to q$ — это **обещание**: «если $p$ (условие), то $q$ (следствие)». Обещание нарушено только если условие выполнено ($p$ истинно), а следствие — нет ($q$ ложно). Во всех остальных случаях обещание держится.

*Юмор от Профессора:* «Политик говорит: "Если выберете меня, построю мост". Его не выбрали — он не построил. Он нарушил обещание? Нет! Условие не выполнено. Он честен (по логике, не по жизни).»

## 7.5. Техническое объяснение

**Импликация** $p \to q$ (читается «если $p$, то $q$» или «$p$ влечёт $q$»):

| $p$ | $q$ | $p \to q$ |
|---|---|---|
| T | T | **T** |
| T | F | **F** | ← единственный случай ложности!
| F | T | **T** |
| F | F | **T** |

**Ложна только если $p$ истинно, а $q$ ложно.** Во всех остальных случаях — истинна.

**Терминология:**
- $p$ — **посылка** (антецедент, условие).
- $q$ — **заключение** (консеквент, следствие).

**Эквивалентные формы:**
- $p \to q \equiv \neg p \lor q$  ← ключевая эквивалентность!
- $p \to q \equiv \neg q \to \neg p$  (контрапозиция)

**Свойства:**
- **Не коммутативна:** $p \to q \not\equiv q \to p$.
- **Не ассоциативна:** $(p \to q) \to r \not\equiv p \to (q \to r)$.
- $\bot \to q \equiv \top$ («из лжи что угодно»).
- $p \to \top \equiv \top$ («истина следует из чего угодно»).

## 7.6. Внутреннее устройство

**Почему $F \to T = T$ и $F \to F = T$?** Это самое сложное. Интуиция: если условие не выполнено ($p = F$), то обещание **автоматически держится** (его нарушить невозможно). Политик не обещал строить мост, если не выбран — он не выбран, обещание держится (что бы он ни делал).

**Почему $p \to q \equiv \neg p \lor q$?** Подставим таблицу:
- $p=T, q=T$: $\neg T \lor T = F \lor T = T$ ✓
- $p=T, q=F$: $\neg T \lor F = F \lor F = F$ ✓
- $p=F, q=T$: $\neg F \lor T = T \lor T = T$ ✓
- $p=F, q=F$: $\neg F \lor F = T \lor F = T$ ✓

Все 4 случая совпадают. Это и есть определение импликации через $\neg$ и $\lor$.

## 7.7. Визуальная схема

```
Импликация p → q — "обещание": ложна только если p=T, q=F

p    q    p→q
T    T     T   ← обещание держится: дождь + зонт
T    F     F   ← обещание НАРУШЕНО: дождь, а зонта нет
F    T     T   ← обещание держится: нет дождя, зонт не обязателен
F    F     T   ← обещание держится: нет дождя, нет зонта

Эквивалентность: p → q ≡ ¬p ∨ q
p    q    ¬p    ¬p∨q
T    T    F     T      ← совпадает с p→q
T    F    F     F
F    T    T     T
F    F    T     T

Контрапозиция: p → q ≡ ¬q → ¬p
"Если дождь, то возьму зонт" ≡ "Если не взял зонт, то дождя не было"

Свойства:
- p → q ≠ q → p                    (не коммутативна!)
- F → q = T всегда                 ("из лжи что угодно")
- p → T = T всегда                 ("истина из чего угодно")
```

## 7.8. Рабочий пример

````python
# ============================================================
# Импликация в Python
# ============================================================

def implies(p, q):
    # Импликация: p → q. Эквивалент ¬p ∨ q.
    return (not p) or q

# Таблица истинности
print(f"{'p':>6} {'q':>6} | {'p→q':>6} | {'¬p':>5} | {'¬p∨q':>6} | совпадает?")
print("-" * 60)
for p in [True, False]:
    for q in [True, False]:
        impl = implies(p, q)
        not_p = not p
        via_basic = not_p or q
        match = impl == via_basic
        print(f"{str(p):>6} {str(q):>6} | {str(impl):>6} | {str(not_p):>5} | {str(via_basic):>6} | {'✓' if match else '✗'}")

# Примеры с обещаниями
print("\n--- Примеры обещаний ---")
promises = [
    ("Если дождь, то возьму зонт", True, False),     # дождь, без зонта — нарушил!
    ("Если дождь, то возьму зонт", True, True),      # дождь, зонт — держит
    ("Если дождь, то возьму зонт", False, True),     # нет дождя, зонт — держит (не нарушал)
    ("Если дождь, то возьму зонт", False, False),    # нет дождя, нет зонта — держит
]
for promise, p, q in promises:
    result = implies(p, q)
    status = "НАРУШЕНО" if not result else "держит"
    print(f"  p={str(p):>5}, q={str(q):>5} → {status}")
````

## 7.9. Практический эксперимент

Проверим контрапозицию: $p \to q \equiv \neg q \to \neg p$

````python
# ============================================================
# Эксперимент: проверка контрапозиции
# ============================================================
import itertools

print("Проверка контрапозиции: p → q ≡ ¬q → ¬p")
print()
print(f"{'p':>6} {'q':>6} | {'p→q':>6} | {'¬q':>5} {'¬p':>5} | {'¬q→¬p':>7} | совпадает?")
print("-" * 65)

all_match = True
for p, q in itertools.product([True, False], repeat=2):
    left = implies(p, q)
    not_q = not q
    not_p = not p
    right = implies(not_q, not_p)
    match = left == right
    if not match:
        all_match = False
    print(f"{str(p):>6} {str(q):>6} | {str(left):>6} | {str(not_q):>5} {str(not_p):>5} | {str(right):>7} | {'✓' if match else '✗'}")

print(f"\nВсе 4 строки совпадают? {all_match}")

# Применение в доказательствах
print("\n--- Применение: доказательство от противного ---")
print("Хотим доказать: 'Если n² чётно, то n чётно'.")
print("Контрапозиция: 'Если n НЕчётно, то n² НЕчётно'.")
print("Доказываем контрапозицию: n=2k+1 → n²=4k²+4k+1 = 2(2k²+2k)+1 — нечётно. ✓")
print("Значит, исходная импликация верна (по контрапозиции).")

# Профессор: "Контрапозиция — мощный приём. Часто доказать ¬q → ¬p проще, чем p → q."
````

$F$. Это единственный случай, когда импликация ложна: посылка истинна, заключение ложно.

**Вопрос 2.** Чему равно $F \to T$?

<details><summary><b>Ответ</b></summary>
$T$. Если посылка ложна, импликация истинна (обещание не нарушено).
</details>

**Вопрос 3.** Чему равно $p \to q$ через $\neg$ и $\lor$?

<details><summary><b>Ответ</b></summary>
$p \to q \equiv \neg p \lor q$.
</details>

**Вопрос 4.** Верна ли коммутативность $p \to q = q \to p$?

<details><summary><b>Ответ</b></summary>
**Нет**. Импликация не коммутативна. $T \to F = F$, но $F \to T = T$.
</details>

## 7.11. Типичные ошибки

**Ошибка 1.** «$F \to T = F$.»
- **Причина:** думают, что ложная посылка даёт ложное следствие.
- **Правильно:** $F \to T = T$. Если посылка ложна, импликация истинна.

**Ошибка 2.** «$p \to q$ значит, что $p$ **причинно** влечёт $q$.»
- **Причина:** переносят разговорный смысл.
- **Правильно:** в логике **нет** причинности. $p \to q$ — только таблица истинности. «Если 2+2=4, то Москва — столица России» — это $T \to T = T$, хотя причинной связи нет.

**Ошибка 3.** «$p \to q = q \to p$.»
- **Причина:** путают с коммутативностью.
- **Правильно:** импликация **не** коммутативна. Только $p \to q \equiv \neg q \to \neg p$ (контрапозиция).

## 7.12. Краткий вывод и переход

- $p \to q$ — импликация, ложна только если $p$ истинно, $q$ ложно.
- $p \to q \equiv \neg p \lor q$.
- $p \to q \equiv \neg q \to \neg p$ (контрапозиция).
- Не коммутативна, не ассоциативна.
- В Python: `(not p) or q`.

**Чек-лист:**

# Тема 8. Эквивалентность и законы де Моргана

## 8.1. Постановка проблемы и практический пример

«Я куплю билет **тогда и только тогда**, когда будет свободное место». Это **эквивалентность**: оба условия истинны одновременно, или оба ложны. Обозначается $\leftrightarrow$.

Также в этой теме — **законы де Моргана**, которые позволяют «выносить» отрицание через $\land$ и $\lor$. Это **самые важные законы** логики, спасающие от ошибок при переписывании условий.

## 8.3. Интуитивное объяснение и аналогия

> **Аналогия:** Эквивалентность $p \leftrightarrow q$ — это **синхронность**: обе лампочки горят или обе не горят. Если одна горит, а другая нет — эквивалентность ложна.

*Юмор от Профессора:* «Законы де Моргана — это "правило развода" в логике: "не (муж ИЛИ жена)" = "не муж И не жена". Август де Морган был математиком, но понимал жизнь.»

## 8.5. Техническое объяснение

**Эквивалентность** $p \leftrightarrow q$ (читается «$p$ тогда и только тогда, когда $q$»):

| $p$ | $q$ | $p \leftrightarrow q$ |
|---|---|---|
| T | T | **T** |
| T | F | F |
| F | T | F |
| F | F | **T** |

**Истинна если $p$ и $q$ одинаковой истинности** (оба T или оба F).

**Эквивалентные формы:** $p \leftrightarrow q \equiv (p \to q) \land (q \to p) \equiv (p \land q) \lor (\neg p \land \neg q)$.

**Законы де Моргана:**
1. $\neg(p \land q) \equiv \neg p \lor \neg q$
2. $\neg(p \lor q) \equiv \neg p \land \neg q$

Интуиция: отрицание конъюнкции = дизъюнкция отрицаний, и наоборот. «Не (И)» = «ИЛИ не», «Не (ИЛИ)» = «И не».

## 8.6. Внутреннее устройство

**Почему законы де Моргана так важны?**
- Они позволяют «протащить» отрицание внутрь скобок.
- В программировании: `if not (a and b)` ↔ `if (not a) or (not b)` — это **та же** проверка.
- В SQL: `WHERE NOT (x > 0 AND y < 10)` ↔ `WHERE x <= 0 OR y >= 10`.

**Обобщение на $n$ переменных:**
- $\neg(p_1 \land p_2 \land \ldots \land p_n) \equiv \neg p_1 \lor \neg p_2 \lor \ldots \lor \neg p_n$
- $\neg(p_1 \lor p_2 \lor \ldots \lor p_n) \equiv \neg p_1 \land \neg p_2 \land \ldots \land \neg p_n$

## 8.7. Визуальная схема

```
Эквивалентность p ↔ q — "синхронность":

p    q    p↔q
T    T     T   ← оба истинны
T    F     F
F    T     F
F    F     T   ← оба ложны

p ↔ q = (p→q) ∧ (q→p) = (p∧q) ∨ (¬p∧¬q)

Законы де Моргана:

¬(p ∧ q) ≡ ¬p ∨ ¬q
"не (И)" = "ИЛИ не"

¬(p ∨ q) ≡ ¬p ∧ ¬q
"не (ИЛИ)" = "И не"

Обобщение на n:
¬(p₁ ∧ p₂ ∧ ... ∧ pₙ) ≡ ¬p₁ ∨ ¬p₂ ∨ ... ∨ ¬pₙ
¬(p₁ ∨ p₂ ∨ ... ∨ pₙ) ≡ ¬p₁ ∧ ¬p₂ ∧ ¬p₃ ∧ ... ∧ ¬pₙ

Приложения:
- not (a and b)        ↔ (not a) or (not b)
- not (a or b)         ↔ (not a) and (not b)
- WHERE NOT (x>0 AND y<10)  ↔ WHERE x<=0 OR y>=10
```

## 8.8. Рабочий пример

````python
# ============================================================
# Эквивалентность и законы де Моргана
# ============================================================

def equiv(p, q):
    # Эквивалентность: p ↔ q. Истинна если оба одинаковой истинности.
    return p == q

def neg(p):
    return not p

def conj(p, q):
    return p and q

def disj(p, q):
    return p or q

# 1. Таблица эквивалентности
print("--- Эквивалентность p ↔ q ---")
print(f"{'p':>6} {'q':>6} | {'p↔q':>6} | {'(p→q)∧(q→p)':>13}")
print("-" * 45)
for p in [True, False]:
    for q in [True, False]:
        e = equiv(p, q)
        via_impl = conj((not p) or q, (not q) or p)  # (p→q) ∧ (q→p)
        print(f"{str(p):>6} {str(q):>6} | {str(e):>6} | {str(via_impl):>13}")

# 2. Проверка законов де Моргана
print("\n--- Закон де Моргана 1: ¬(p ∧ q) ≡ ¬p ∨ ¬q ---")
print(f"{'p':>6} {'q':>6} | {'¬(p∧q)':>8} | {'¬p∨¬q':>7} | совпадает?")
print("-" * 50)
all_ok = True
for p in [True, False]:
    for q in [True, False]:
        left = neg(conj(p, q))
        right = disj(neg(p), neg(q))
        match = left == right
        if not match: all_ok = False
        print(f"{str(p):>6} {str(q):>6} | {str(left):>8} | {str(right):>7} | {'✓' if match else '✗'}")
print(f"Все совпадают? {all_ok}")

print("\n--- Закон де Моргана 2: ¬(p ∨ q) ≡ ¬p ∧ ¬q ---")
all_ok = True
for p in [True, False]:
    for q in [True, False]:
        left = neg(disj(p, q))
        right = conj(neg(p), neg(q))
        match = left == right
        if not match: all_ok = False
print(f"Все 4 строки совпадают? {all_ok}")
````

## 8.9. Практический эксперимент

Проверим обобщение де Моргана на 3 переменных:

````python
# ============================================================
# Эксперимент: де Морган для 3 переменных
# ============================================================
import itertools

print("Закон де Моргана для 3 переменных:")
print("¬(p ∧ q ∧ r) ≡ ¬p ∨ ¬q ∨ ¬r")
print()
print(f"{'p':>6} {'q':>6} {'r':>6} | {'p∧q∧r':>6} | {'¬(p∧q∧r)':>10} | {'¬p∨¬q∨¬r':>9} | совпадает?")
print("-" * 70)

all_match = True
for p, q, r in itertools.product([True, False], repeat=3):
    conj3 = p and q and r
    left = not conj3
    right = (not p) or (not q) or (not r)
    match = left == right
    if not match: all_match = False
    print(f"{str(p):>6} {str(q):>6} {str(r):>6} | {str(conj3):>6} | {str(left):>10} | {str(right):>9} | {'✓' if match else '✗'}")

print(f"\nВсе 8 строк совпадают? {all_match}")

# Применение в Python: переписываем условия
print("\n--- Применение: переписываем NOT (a AND b) ---")
def test_data():
    return [(True, True), (True, False), (False, True), (False, False)]

print(f"{'a':>6} {'b':>6} | {'not (a and b)':>15} | {'(not a) or (not b)':>20} | совпадает?")
print("-" * 70)
for a, b in test_data():
    v1 = not (a and b)
    v2 = (not a) or (not b)
    print(f"{str(a):>6} {str(b):>6} | {str(v1):>15} | {str(v2):>20} | {'✓' if v1==v2 else '✗'}")

# Профессор: "Это полезно при рефакторинге. 'not (a and b)' можно переписать как '(not a) or (not b)'."
````

# Тема 9. Предикаты

## 9.1. Постановка проблемы и практический пример

«$x > 5$» — это **не** высказывание (нельзя сказать True/False, не зная $x$). Но если подставить $x = 10$, получится True. Если $x = 3$ — False.

Это **предикат** — высказывание **с параметром**. Обозначается $P(x), Q(x, y), \ldots$

## 9.3. Интуитивное объяснение и аналогия

> **Аналогия:** Предикат — это **форма для печати**: сама по себе не True и не False, но если «заполнить» (подставить значение), получится конкретное высказывание. $P(x) = (x > 5)$ — форма. Подставили $x = 10$ — получили «$10 > 5$» = True.

*Юмор от Профессора:* «Предикат — это резюме программиста без указания зарплаты. Само по себе не True и не False, но подставь конкретную цифру — и сразу понятно, подходит он тебе или нет.»

## 9.5. Техническое объяснение

**Предикат** $P(x)$ — это функция, которая принимает значение (или несколько) и возвращает True или False.

$$P: D \to \{\text{True}, \text{False}\}$$

где $D$ — **область определения** (множество возможных значений $x$).

**Примеры:**
- $P(x) = (x > 0)$ на $D = \mathbb{R}$. Подстановка $x = 5$ даёт $P(5) = T$.
- $Q(n) = (n \text{ чётное})$ на $D = \mathbb{Z}$. $Q(4) = T$, $Q(7) = F$.
- $R(x, y) = (x + y = 0)$ на $D = \mathbb{R} \times \mathbb{R}$. $R(3, -3) = T$.

**Связь с высказываниями:**
- Предикат без параметров = высказывание.
- Подстановка значения в предикат = высказывание.

**Операции над предикатами:** те же, что и над высказываниями ($\neg, \land, \lor, \to, \leftrightarrow$), применяются **поточечно**.

## 9.6. Внутреннее устройство

**Множество истинности предиката** $P(x)$:
$$I_P = \{x \in D : P(x) = \text{True}\}$$

Это подмножество $D$, на котором предикат истинен.

**Пример.** Для $P(x) = (x^2 = x)$ на $D = \mathbb{R}$:
$$I_P = \{0, 1\}$$
(только $0^2 = 0$ и $1^2 = 1$; для других $x$ равенство не выполнено).

**Применение в программировании:**
- Фильтры: `df[df.x > 0]` — берёт строки, где предикат `x > 0` истинен.
- Лямбды: `lambda x: x % 2 == 0` — предикат чётности.
- SQL `WHERE` — предикат.

## 9.7. Визуальная схема

```
Предикат P(x) = (x > 5) на D = R

         P(x)
         ^
         |  True для x > 5
---------+---------> x
         |   5
         |  False для x ≤ 5

Множество истинности: I_P = {x : x > 5} = (5, +∞)

P(x) = (x² = x) на D = R
         I_P = {0, 1}

P(x, y) = (x + y = 0) на D = R × R
         I_P = {(x, -x) : x ∈ R}  — прямая y = -x

Операции:
- ¬P(x)        — истинно где P ложно
- P(x) ∧ Q(x)  — истинно где оба истинны
- P(x) ∨ Q(x)  — истинно где хотя бы один
- P(x) → Q(x)  — истинно где (P ложен) или (Q истинен)
```

## 9.8. Рабочий пример

````python
# ============================================================
# Предикаты в Python
# ============================================================

# Предикат P(x): x² = x  (истинен только для x=0 и x=1)
def P(x):
    return x * x == x                    # сравниваем x² и x

# Проверим на разных x
print("P(x) = (x² == x):")
test_values = [-2, -1, 0, 0.5, 1, 2, 3, 10]
for x in test_values:
    print(f"  P({x}) = {P(x)}")

# Множество истинности на целых числах [-5, 5]
D = list(range(-5, 6))
I_P = [x for x in D if P(x)]
print(f"\nНа D = {D}, I_P = {I_P}  (только 0 и 1)")

# Предикат от двух переменных: Q(x, y) = (x + y == 0)
def Q(x, y):
    return x + y == 0

print(f"\nQ(x, y) = (x + y == 0):")
test_pairs = [(3, -3), (5, 5), (0, 0), (1, 2), (-7, 7)]
for x, y in test_pairs:
    print(f"  Q({x}, {y}) = {Q(x, y)}")

# Операции над предикатами
def D_pred(x):                          # "x делится на 2"
    return x % 2 == 0

def E_pred(x):                          # "x делится на 3"
    return x % 3 == 0

print(f"\nОперации над предикатами на x ∈ [0..9]:")
print(f"{'x':>4} | {'D (÷2)':>8} {'E (÷3)':>8} {'¬D':>6} {'D∧E':>6} {'D∨E':>6} {'D→E':>6}")
print("-" * 55)
for x in range(10):
    d = D_pred(x)
    e = E_pred(x)
    not_d = not d
    d_and_e = d and e
    d_or_e = d or e
    d_impl_e = (not d) or e
    print(f"{x:>4} | {str(d):>8} {str(e):>8} {str(not_d):>6} {str(d_and_e):>6} {str(d_or_e):>6} {str(d_impl_e):>6}")
````

````python
# ============================================================
# Эксперимент: предикаты в pandas-стиле фильтрации
# ============================================================

# Список транзакций: (id, сумма, категория, одобрена)
transactions = [
    (1,  1500,  "еда",     True),
    (2,  50,    "транспорт", True),
    (3,  30000, "развлечения", False),
    (4,  850,   "еда",     True),
    (5,  5000,  "техника", True),
    (6,  200,   "транспорт", False),
    (7,  12000, "техника", True),
]

# Предикаты
def is_large(t):           # сумма > 1000
    return t[1] > 1000

def is_food(t):            # категория = еда
    return t[2] == "еда"

def is_approved(t):        # одобрена
    return t[3]

# Применяем предикаты
print("Все транзакции:")
for t in transactions:
    print(f"  {t}")

print(f"\nКрупные (сумма > 1000): {sum(is_large(t) for t in transactions)}")
for t in transactions:
    if is_large(t):
        print(f"  ✓ {t}")

print(f"\nЕда: {sum(is_food(t) for t in transactions)}")
for t in transactions:
    if is_food(t):
        print(f"  ✓ {t}")

# Композиция предикатов
print(f"\nКрупная еда (is_large ∧ is_food): {sum(is_large(t) and is_food(t) for t in transactions)}")
print(f"Одобрено ИЛИ крупная (is_approved ∨ is_large): {sum(is_approved(t) or is_large(t) for t in transactions)}")
print(f"Не еда И не транспорт (¬is_food ∧ ¬is_transport): {sum(not is_food(t) and t[2] != 'транспорт' for t in transactions)}")

# Профессор: "Любой WHERE в SQL — это предикат. Любой if в Python — применение предиката."
````

# Тема 10. Квантор $\forall$ (всеобщность)

## 10.1. Постановка проблемы и практический пример

«Все ученики в классе сдали экзамен». Это **высказывание** (можно сказать True/False), но оно говорит о **всех** элементах множества. Его можно записать через квантор:

$$\forall x \in \text{Класс}: \text{Сдал}(x)$$

Читается: «для всех $x$ из Класса верно, что $x$ сдал экзамен». Это **квантор всеобщности** $\forall$.

## 10.3. Интуитивное объяснение и аналогия

> **Аналогия:** $\forall$ — это **тщательный ревизор**. Он проверяет **каждого**. Если хотя бы один не сдал — ревизор ставит False. Все сдали — True. Ревизор не успокаивается, пока не проверит всех.

*Юмор от Профессора:* «$\forall$ — строгий учитель. Один ученик не сделал домашку — и весь класс лишается пиццы. Жестоко, но математично.»

## 10.5. Техническое объяснение

**Квантор всеобщности** $\forall x \in D: P(x)$ (читается «для всех $x$ из $D$ выполняется $P(x)$»).

**Определение:**
$$\forall x \in D: P(x) = \begin{cases} \text{True}, & \text{если } P(x) = \text{True} \text{ для всех } x \in D \\ \text{False}, & \text{если найдётся } x \in D \text{ с } P(x) = \text{False} \end{cases}$$

**Контрпример.** Чтобы доказать $\forall x: P(x)$ ложен, достаточно **одного** $x$ с $P(x) = F$.

**В Python:** `all(P(x) for x in D)`.

**Примеры:**
- $\forall x \in \mathbb{R}: x^2 \geq 0$ — True (квадрат всегда неотрицателен).
- $\forall x \in \mathbb{R}: x > 0$ — False (например, $x = -1$).
- $\forall n \in \mathbb{N}: n + 1 > n$ — True.

## 10.6. Внутреннее устройство

**Связь с конъюнкцией.** Для **конечного** $D = \{a_1, a_2, \ldots, a_n\}$:
$$\forall x \in D: P(x) \equiv P(a_1) \land P(a_2) \land \ldots \land P(a_n)$$

То есть $\forall$ = «И для всех». Если $D$ бесконечно — $\forall$ обобщает конъюнкцию на бесконечный случай.

**Логика квантора:**
- $\forall x: P(x) = T$ — все элементы удовлетворяют.
- $\forall x: P(x) = F$ — **хотя бы один** не удовлетворяет (есть контрпример).

## 10.7. Визуальная схема

```
Квантор ∀ (всеобщность): "для всех"

∀x ∈ D: P(x) = True если P истинно для КАЖДОГО x ∈ D

                  P(x)
                   ^
              True |  • • • • • • • • •     ← все элементы P истинны
                   |
            --------+--------> x   (D = вся прямая)
                   |
             False |
                  
                  → ∀x: P(x) = T

Если есть контрпример:
                  P(x)
                   ^
              True |  • • •   • • • •
                   |        ✗          ← один элемент, где P ложно
            --------+--------> x
                   |
             False |
                  → ∀x: P(x) = F

Python: all(P(x) for x in D)
SQL:    WHERE ALL (SELECT ... )
```

## 10.8. Рабочий пример

````python
# ============================================================
# Квантор ∀ в Python
# ============================================================

# Предикат P(x): x² ≥ 0  (всегда истинно для вещественных)
def P(x):
    return x * x >= 0

# Проверим ∀x ∈ [-5, 5]: P(x)
D = list(range(-5, 6))
result = all(P(x) for x in D)            # квантор ∀ через all()
print(f"∀x ∈ {D}: x² ≥ 0  →  {result}")

# Предикат Q(x): x > 0
def Q(x):
    return x > 0

# ∀x ∈ [-5, 5]: Q(x)  — должно быть False (есть x ≤ 0)
result = all(Q(x) for x in D)
print(f"∀x ∈ {D}: x > 0  →  {result}")

# Найдём контрпример
counter = next((x for x in D if not Q(x)), None)
print(f"Контрпример: x = {counter} (Q({counter}) = {Q(counter)})")

# Предикат R(n): n + 1 > n
def R(n):
    return n + 1 > n

# ∀n ∈ [0, 1000]: R(n)
D2 = list(range(0, 1001))
result = all(R(n) for n in D2)
print(f"\n∀n ∈ [0..1000]: n+1 > n  →  {result}")
print(f"(можно проверить и для всего Z — это всегда True)")

# Сравним с квантором существования
print(f"\n∃n ∈ [0..1000]: n² = 25 → {any(n*n == 25 for n in D2)}")
````

## 10.9. Практический эксперимент

Применение $\forall$ для верификации:

````python
# ============================================================
# Эксперимент: ∀ для проверки инвариантов
# ============================================================

# Применение: проверка инвариантов цикла
# "Все элементы отсортированного списка идут по возрастанию"
def is_sorted(lst):
    # ∀i: lst[i] ≤ lst[i+1]
    return all(lst[i] <= lst[i+1] for i in range(len(lst) - 1))

# Тесты
tests = [
    ([1, 2, 3, 4, 5], True),
    ([1, 1, 1, 1],    True),       # равные — ок
    ([5, 4, 3, 2, 1], False),
    ([1, 2, 1, 2],    False),
    ([],              True),       # пустой — тривиально отсортирован
    ([42],            True),       # один элемент — отсортирован
]

print(f"{'Список':<22} | {'Отсортирован?':>14} | {'Ожидание':>10}")
print("-" * 55)
for lst, expected in tests:
    result = is_sorted(lst)
    marker = "✓" if result == expected else "✗"
    print(f"{str(lst):<22} | {str(result):>14} | {str(expected):>10} {marker}")

# Применение: проверка, что ВСЕ студенты сдали
students = {"Анна": 85, "Боря": 72, "Вася": 90, "Галя": 65, "Дима": 88}
passing_grade = 60

all_passed = all(grade >= passing_grade for grade in students.values())
print(f"\nСтуденты и оценки: {students}")
print(f"∀студент: оценка ≥ {passing_grade}  →  {all_passed}")

# Профессор: "all() в Python — это и есть ∀. any() — это ∃. Запомни на всю жизнь."
````

# Тема 11. Квантор $\exists$ (существование)

## 11.1. Постановка проблемы и практический пример

«В классе **есть** ученик, который получил 100 баллов». Это высказывание говорит о **существовании** хотя бы одного такого ученика. Записывается:

$$\exists x \in \text{Класс}: \text{Балл}(x) = 100$$

Читается: «существует $x$ из Класса, для которого балл $x$ равен 100». Это **квантор существования** $\exists$.

## 11.3. Интуитивное объяснение и аналогия

> **Аналогия:** $\exists$ — это **поиск сокровищ**. Ищешь хотя бы один сундук с золотом. Нашёл — True. Не нашёл (хотя бы одного) — False. Не нужно проверять все сундуки — достаточно первого найденного.

*Юмор от Профессора:* «$\exists$ — оптимист среди кванторов. "Хотя бы один!" — его девиз. В отличие от $\forall$ — пессимиста, который верит только если абсолютно все соответствует.»

## 11.5. Техническое объяснение

**Квантор существования** $\exists x \in D: P(x)$ (читается «существует $x$ из $D$ такой, что $P(x)$»).

**Определение:**
$$\exists x \in D: P(x) = \begin{cases} \text{True}, & \text{найдётся } x \in D \text{ с } P(x) = T \\ \text{False}, & \text{если } P(x) = F \text{ для всех } x \in D \end{cases}$$

**Свидетель.** Чтобы доказать $\exists x: P(x)$ истинно, достаточно **привести один пример** $x$ с $P(x) = T$.

**В Python:** `any(P(x) for x in D)`.

**Примеры:**
- $\exists x \in \mathbb{R}: x^2 = 4$ — True (например, $x = 2$).
- $\exists x \in \mathbb{Z}: x^2 = 2$ — False (нет целого корня из 2).
- $\exists x \in \mathbb{N}: x > 100$ — True (например, $x = 101$).

## 11.6. Внутреннее устройство

**Связь с дизъюнкцией.** Для **конечного** $D = \{a_1, \ldots, a_n\}$:
$$\exists x \in D: P(x) \equiv P(a_1) \lor P(a_2) \lor \ldots \lor P(a_n)$$

То есть $\exists$ = «ИЛИ для некоторых». Обобщает дизъюнкцию на бесконечный случай.

**Логика квантора:**
- $\exists x: P(x) = T$ — есть **хотя бы один** удовлетворяющий.
- $\exists x: P(x) = F$ — **все** не удовлетворяют.

**Связь с $\forall$:** $\exists x: P(x) \equiv \neg(\forall x: \neg P(x))$. То есть «существует удовлетворяющий» = «неверно, что все не удовлетворяют».

## 11.7. Визуальная схема

```
Квантор ∃ (существование): "существует хотя бы один"

∃x ∈ D: P(x) = True если найдётся x ∈ D с P(x) = True

                  P(x)
                   ^
              True |  • •   •   •   •     ← достаточно одного True
                   |
            --------+--------> x
                   |
             False |

                  → ∃x: P(x) = T (есть зелёные точки)

Если все False:
              True |  (никаких точек)
            --------+--------> x
             False |  • • • • • • • • • •
                  → ∃x: P(x) = F

Python: any(P(x) for x in D)
SQL:    WHERE EXISTS (SELECT ...)

Связь: ∃x: P(x) ≡ ¬(∀x: ¬P(x))
       "существует" ≡ "неверно, что все не удовлетворяют"
```

## 11.8. Рабочий пример

````python
# ============================================================
# Квантор ∃ в Python
# ============================================================

# Предикат D(x): x делится на 2 и на 3 одновременно (т.е. на 6)
def D(x):
    return x % 2 == 0 and x % 3 == 0

# Предикат E(x): x делится на 6 (то же самое)
def E(x):
    return x % 6 == 0

# Множество M = [0..9]
M = list(range(0, 10))

# ∃x ∈ M: D(x) — есть ли число, делящееся и на 2, и на 3?
exists_d = any(D(x) for x in M)
print(f"∃x ∈ {M}: (x÷2 ∧ x÷3)  →  {exists_d}")

# Найти свидетель (пример)
witness = next((x for x in M if D(x)), None)
print(f"Свидетель: x = {witness} (D({witness}) = {D(witness)})")

# ∀x ∈ M: D(x) — все ли числа делятся и на 2, и на 3?
all_d = all(D(x) for x in M)
print(f"∀x ∈ {M}: (x÷2 ∧ x÷3)  →  {all_d}")

# Сложный предикат: ∃x ∈ Z: x² = 2
# Проверим на диапазоне целых чисел
def square_is_2(x):
    return x * x == 2

Z_range = list(range(-100, 101))
exists_sqrt2 = any(square_is_2(x) for x in Z_range)
print(f"\n∃x ∈ [-100, 100] ∩ Z: x² = 2  →  {exists_sqrt2}")
print(f"Это значит: целого корня из 2 НЕТ (контр-интуитивно, но факт).")

# ∃x ∈ R: x² = 2 — есть (x = √2), но мы не можем перебрать R
# В Python проверим только приближённо
import math
print(f"\n∃x ∈ R: x² = 2  →  True, x = √2 ≈ {math.sqrt(2):.6f}")
print(f"Но √2 ∉ Z, поэтому ∃x ∈ Z: x² = 2  →  False")
````

## 11.9. Практический эксперимент

Визуализируем $\forall$ и $\exists$:

````python
# ============================================================
# Эксперимент: визуализация ∀ и ∃
# ============================================================
import matplotlib.pyplot as plt
import numpy as np

# 4 предиката на множестве M = [0..9]
M = np.arange(0, 10)

predicates = [
    ("x² == x (только 0 и 1)", lambda x: x*x == x),
    ("x чётное",                lambda x: x % 2 == 0),
    ("x > 5",                   lambda x: x > 5),
    ("x > 10 (никто)",          lambda x: x > 10),
]

fig, axes = plt.subplots(2, 2, figsize=(13, 8), constrained_layout=True)

for ax, (name, pred) in zip(axes.flat, predicates):
    values = [pred(x) for x in M]
    colors = ['#27ae60' if v else '#e74c3c' for v in values]
    ax.bar(M, [1]*10, color=colors, edgecolor='black', linewidth=1.2)
    
    # Маркируем зелёные (True) и красные (False)
    for i, v in enumerate(values):
        ax.text(i, 0.5, 'T' if v else 'F', ha='center', va='center',
                fontsize=14, fontweight='bold', color='white')
    
    # Считаем ∀ и ∃
    forall = all(values)
    exists = any(values)
    
    ax.set_title(f'{name}\n∀: {"T" if forall else "F"}    ∃: {"T" if exists else "F"}', fontsize=11)
    ax.set_xticks(M)
    ax.set_yticks([])
    ax.set_ylim(0, 1.1)

plt.suptitle('Кванторы ∀ и ∃ на 4 предикатах', fontsize=14, y=1.02)
plt.show()

# Профессор: "Зелёные точки — там, где предикат истинен."
#             "∀ = T если все зелёные. ∃ = T если хотя бы одна зелёная."
````

# Тема 12. Отрицание кванторов

## 12.1. Постановка проблемы и практический пример

«Неверно, что все сдали экзамен». Это то же самое, что «хотя бы один не сдал». То есть:

$$\neg(\forall x: P(x)) \equiv \exists x: \neg P(x)$$

Аналогично, «неверно, что хотя бы один сдал» = «никто не сдал» = «все не сдали»:

$$\neg(\exists x: P(x)) \equiv \forall x: \neg P(x)$$

Это **законы де Моргана для кванторов**. Они **переворачивают** квантор: $\forall \leftrightarrow \exists$.

## 12.3. Интуитивное объяснение и аналогия

> **Аналогия:** Как и обычные законы де Моргана, только для «все»/«существует». «Не все» = «некоторые не». «Никто не» = «все не». Квантор **меняется** на парный.

*Юмор от Профессора:* «Если профессор говорит "не все сдали", это значит "хотя бы один не сдал". Логика спасает от недопонимания. Хотя на практике студент идёт пересдавать в любом случае.»

## 12.5. Техническое объяснение

**Законы де Моргана для кванторов:**

1. $\neg(\forall x: P(x)) \equiv \exists x: \neg P(x)$
2. $\neg(\exists x: P(x)) \equiv \forall x: \neg P(x)$

**Интуиция:**
- «Не все» = «существует хотя бы один, который не».
- «Не существует» = «все не».

**Мнемоника:** отрицание **меняет** $\forall \leftrightarrow \exists$ и **протаскивает** $\neg$ внутрь.

## 12.6. Внутреннее устройство

**Почему это работает?**
- $\forall x: P(x)$ = $P(a_1) \land P(a_2) \land \ldots$ — конъюнкция всех.
- $\neg(\forall x: P(x)) = \neg(P(a_1) \land P(a_2) \land \ldots) = \neg P(a_1) \lor \neg P(a_2) \lor \ldots$ (по обычному де Моргану) $= \exists x: \neg P(x)$.

Аналогично для $\exists$.

**Применения:**
- В математике: чтобы доказать «не для всех», ищут **контрпример**.
- В SQL: `NOT EXISTS (SELECT ...)` = «все не удовлетворяют».
- В Python: `not all(P(x) for x in D)` = `any(not P(x) for x in D)`.

## 12.7. Визуальная схема

```
Законы де Моргана для кванторов:

1. ¬(∀x: P(x)) ≡ ∃x: ¬P(x)
   "не все" = "существует хотя бы один, который не"

2. ¬(∃x: P(x)) ≡ ∀x: ¬P(x)
   "не существует" = "все не"

Мнемоника: ¬ меняет ∀ ↔ ∃, и протаскивает ¬ внутрь.

Примеры:
- ¬(∀x: x² > 0)  ≡  ∃x: x² ≤ 0     (это x=0)
- ¬(∃x ∈ Z: x² = 2)  ≡  ∀x ∈ Z: x² ≠ 2

Связь с обычным де Морганом (для конечных множеств):
¬(P(a₁) ∧ P(a₂) ∧ ...)  ≡  ¬P(a₁) ∨ ¬P(a₂) ∨ ...
¬(∀x: P(x))              ≡  ∃x: ¬P(x)

Python:
- not all(P(x) for x in D)  =  any(not P(x) for x in D)
- not any(P(x) for x in D)  =  all(not P(x) for x in D)
```

## 12.8. Рабочий пример

````python
# ============================================================
# Отрицание кванторов в Python
# ============================================================

# Множество M = [0..9]
M = list(range(0, 10))

# Предикаты
def even(x):  return x % 2 == 0
def positive(x): return x > 0

print("--- Закон 1: ¬(∀x: P(x)) ≡ ∃x: ¬P(x) ---")
print("Проверим на P(x) = (x > 0): ∀x ∈ [0..9]: x > 0  — должно быть False (x=0)")
print()

# Левая часть: ¬(∀x: P(x))
left = not all(positive(x) for x in M)
# Правая часть: ∃x: ¬P(x)
right = any(not positive(x) for x in M)

print(f"M = {M}")
print(f"Левая часть: ¬(∀x: x > 0) = not all(x > 0) = {left}")
print(f"Правая часть: ∃x: x ≤ 0  = any(x ≤ 0) = {right}")
print(f"Совпадают? {left == right}")

print("\n--- Закон 2: ¬(∃x: P(x)) ≡ ∀x: ¬P(x) ---")
print("Проверим на P(x) = (x > 100): ∃x ∈ [0..9]: x > 100  — должно быть False (все ≤ 9)")
print()

# Левая часть: ¬(∃x: P(x))
left = not any(x > 100 for x in M)
# Правая часть: ∀x: ¬P(x)
right = all(not (x > 100) for x in M)

print(f"M = {M}")
print(f"Левая часть: ¬(∃x: x > 100) = not any(x > 100) = {left}")
print(f"Правая часть: ∀x: x ≤ 100 = all(x ≤ 100) = {right}")
print(f"Совпадают? {left == right}")

# Несколько примеров с математикой
print("\n--- Математические примеры ---")
examples = [
    ("¬(∀x ∈ R: x² > 0)", "∃x ∈ R: x² ≤ 0",   "x = 0 (0² = 0, не > 0)"),
    ("¬(∃x ∈ Z: x² = 2)", "∀x ∈ Z: x² ≠ 2",    "√2 не целое, поэтому квадрат целого ≠ 2"),
    ("¬(∀x ∈ N: x чётное)", "∃x ∈ N: x нечётное", "x = 1 (или 3, 5, ...)"),
    ("¬(∃x ∈ R: x² < 0)", "∀x ∈ R: x² ≥ 0",     "квадрат вещественного всегда ≥ 0"),
]
print(f"{'Левая часть':<30} | {'Правая часть':<25} | Истинность/объяснение")
print("-" * 95)
for left_str, right_str, expl in examples:
    print(f"{left_str:<30} | {right_str:<25} | {expl}")
````

# Тема 13. Несколько кванторов

## 13.1. Постановка проблемы и практический пример

Определение предела последовательности: $a_n \to L$ если

$$\forall \varepsilon > 0 \; \exists N \in \mathbb{N}: \forall n > N: |a_n - L| < \varepsilon$$

Здесь **три квантора** подряд: $\forall \exists \forall$. И порядок **важен**: нельзя менять $\forall \varepsilon \exists N$ на $\exists N \forall \varepsilon$ — это **другое** утверждение.

## 13.3. Интуитивное объяснение и аналогия

> **Аналогия:** $\forall x \exists y: P(x, y)$ = «для каждого $x$ найдётся свой $y$» (как **персональный тренер**: каждому клиенту — свой тренер).
>
> $\exists y \forall x: P(x, y)$ = «существует один $y$, который работает для всех $x$» (как **универсальный тренер**: один для всех).
>
> Второе **сильнее**: если есть универсальный $y$, то для каждого $x$ есть $y$ (тот же самый). Но из первого второе не следует!

*Юмор от Профессора:* «$\exists y \forall x$ — это как одинлекарь для всех болезней. $\forall x \exists y$ — у каждого пациента свой врач. Первое — утопия, второе — реальная медицина.»

## 13.5. Техническое объяснение

**Порядок кванторов важен:**

| Утверждение | Что значит |
|---|---|
| $\forall x \exists y: P(x,y)$ | для каждого $x$ есть свой $y$ (зависит от $x$) |
| $\exists y \forall x: P(x,y)$ | есть один $y$ для всех $x$ (не зависит от $x$) |

**Импликация:** $\exists y \forall x: P(x,y) \Rightarrow \forall x \exists y: P(x,y)$ (если есть универсальный, то для каждого есть). **Обратное неверно.**

**Пример 1 (где $\forall \exists$ истинно, а $\exists \forall$ ложно):**
- $P(x, y) = (x + y = 0)$ на $\mathbb{R}$.
- $\forall x \exists y: x + y = 0$ — **True**: для каждого $x$ берём $y = -x$.
- $\exists y \forall x: x + y = 0$ — **False**: нет универсального $y$, который в сумме с любым $x$ даёт 0 (если $y = -x_0$, то для $x_1 \neq x_0$ не сработает).

**Пример 2 (определение предела):**
- $\forall \varepsilon > 0 \exists N: \forall n > N: |a_n - L| < \varepsilon$
- Это значит: для **каждой** точности $\varepsilon$ найдётся **свой** порог $N$ (зависящий от $\varepsilon$).
- Если бы было $\exists N \forall \varepsilon$ — это значило бы, что есть **один** $N$ для всех точностей. Это **сильнее** и почти всегда ложно.

## 13.6. Внутреннее устройство

**Как читать формулы с несколькими кванторами?**
- Слева направо.
- Каждый следующий квантор **может зависеть** от переменных, упомянутых ранее.

**Пример:** $\forall x \exists y \forall z: P(x, y, z)$
- Читаем: «для всех $x$ существует $y$ (который может зависеть от $x$) такой, что для всех $z$ выполняется $P(x, y, z)$».
- $y$ может зависеть от $x$, но не от $z$.

**Отрицание нескольких кванторов:**
$$\neg(\forall x \exists y: P) \equiv \exists x \forall y: \neg P$$
Каждый квантор меняется на парный, и $\neg$ протаскивается в конец.

## 13.7. Визуальная схема

```
Порядок кванторов ВАЖЕН:

∀x ∃y: P(x, y)  ≠  ∃y ∀x: P(x, y)

∀x ∃y: P(x, y) = "для каждого x есть свой y"
                  x₁ → y₁
                  x₂ → y₂
                  x₃ → y₃
                  (y зависит от x)

∃y ∀x: P(x, y) = "есть один y для всех x"
                  x₁ ↘
                  x₂ → y*
                  x₃ ↗
                  (y не зависит от x — СИЛЬНЕЕ)

∃y ∀x: P  ⇒  ∀x ∃y: P    (обратное НЕ верно!)

Пример (P(x, y) = (x + y = 0)):
∀x ∃y: x + y = 0    → True (y = -x, зависит от x)
∃y ∀x: x + y = 0    → False (нет универсального y)

Отрицание нескольких кванторов:
¬(∀x ∃y: P)  ≡  ∃x ∀y: ¬P
"не для каждого x есть y"  =  "есть x, для которого ни один y не подходит"
```

## 13.8. Рабочий пример

````python
# ============================================================
# Несколько кванторов: ∀x ∃y: P(x,y) vs ∃y ∀x: P(x,y)
# ============================================================

# Множество M = [-5, 5]
M = list(range(-5, 6))

# Предикат P(x, y) = (x + y == 0)
def P(x, y):
    return x + y == 0

# 1. ∀x ∃y: P(x, y) — для каждого x найдётся y = -x
forall_exists = all(any(P(x, y) for y in M) for x in M)
print(f"∀x ∈ {M}: ∃y ∈ {M}: x + y = 0  →  {forall_exists}")
print(f"  (для каждого x берём y = -x, который тоже в M)")

# 2. ∃y ∀x: P(x, y) — есть один y, работающий для всех x
exists_forall = any(all(P(x, y) for x in M) for y in M)
print(f"\n∃y ∈ {M}: ∀x ∈ {M}: x + y = 0  →  {exists_forall}")
print(f"  (нет универсального y: если y = -x₁, то для x₂ ≠ x₁ не сработает)")

# Вывод
print(f"\nВывод: ∀∃ = {forall_exists}, ∃∀ = {exists_forall}")
print(f"Из ∃∀ следует ∀∃, но не наоборот — этот пример показывает.")

# Профессор: "P(x,y) = (x + y = 0) — классический пример, где ∀x∃y ≠ ∃y∀x."
#             "Запомни: порядок кванторов ВАЖЕН."
````

## 13.9. Практический эксперимент

Определение предела последовательности $a_n = 1/n \to 0$:

````python
# ============================================================
# Эксперимент: предел последовательности 1/n → 0
# ============================================================
# Определение: ∀ε > 0 ∃N ∈ N: ∀n > N: |a_n - L| < ε
# Где a_n = 1/n, L = 0.

def a(n):
    return 1 / n

L = 0

# Проверим: для разных ε найдём N
epsilons = [0.1, 0.01, 0.001, 0.0001]

print("Предел 1/n → 0:")
print(f"{'ε':>10} | {'N (минимальное)':>15} | {'|a_(N+1) - L|':>15} | {'< ε?':>6}")
print("-" * 60)

for eps in epsilons:
    # Ищем минимальное N такое, что для всех n > N: |1/n - 0| < eps
    # |1/n| < eps  ⟺  n > 1/eps
    N = int(1 / eps)  # примерное значение
    # Точная проверка
    while abs(a(N + 1) - L) >= eps:
        N += 1
    # Проверим, что для всех n > N выполняется
    check_n = N + 1
    check = abs(a(check_n) - L) < eps
    print(f"{eps:>10.4f} | {N:>15} | {abs(a(N+1)-L):>15.6f} | {'✓' if check else '✗':>6}")

print(f"\nДля каждого ε нашли свой N. N зависит от ε — поэтому ∀ε ∃N, а не ∃N ∀ε.")
print(f"Если бы было ∃N ∀ε — это значило бы, что есть один N для всех ε.")
print(f"Но при ε → 0 число N должно расти до бесконечности, так что универсального N нет.")

# Профессор: "Предел 1/n → 0 — классика. Запомни: ∀ε ∃N, не наоборот!"
````

## 13.11. Типичные ошибки

## 13.12. Краткий вывод и переход

## 5 основных операций

| Операция | Символ | Python | Читается | Таблица |
|---|---|---|---|---|
| Отрицание | $\neg p$ | `not p` | «не $p$» | $T \to F, F \to T$ |
| Конъюнкция | $p \land q$ | `p and q` | «$p$ и $q$» | $T$ только если оба $T$ |
| Дизъюнкция | $p \lor q$ | `p or q` | «$p$ или $q$» | $T$ если хотя бы один $T$ |
| XOR | $p \oplus q$ | `p ^ q` или `p != q` | «либо $p$, либо $q$» | $T$ если ровно один $T$ |
| Импликация | $p \to q$ | `(not p) or q` | «если $p$, то $q$» | $F$ только если $p=T, q=F$ |
| Эквивалентность | $p \leftrightarrow q$ | `p == q` | «$p$ тогда и только тогда когда $q$» | $T$ если одинаковой истинности |

## Законы де Моргана

| Что | Формула |
|---|---|
| Не (И) | $\neg(p \land q) \equiv \neg p \lor \neg q$ |
| Не (ИЛИ) | $\neg(p \lor q) \equiv \neg p \land \neg q$ |
| Не (для всех) | $\neg(\forall x: P) \equiv \exists x: \neg P$ |
| Не (существует) | $\neg(\exists x: P) \equiv \forall x: \neg P$ |

## Кванторы

| Квантор | Python | Что значит |
|---|---|---|
| $\forall x: P(x)$ | `all(P(x) for x in D)` | для всех $x$ верно $P$ |
| $\exists x: P(x)$ | `any(P(x) for x in D)` | существует $x$ с $P$ |

## Ключевые эквивалентности

| Что | Формула |
|---|---|
| Импликация | $p \to q \equiv \neg p \lor q$ |
| Контрапозиция | $p \to q \equiv \neg q \to \neg p$ |
| Эквивалентность | $p \leftrightarrow q \equiv (p \to q) \land (q \to p)$ |
| Двойное отрицание | $\neg \neg p \equiv p$ |
| Исключённое третье | $p \lor \neg p \equiv \top$ |
| Противоречие | $p \land \neg p \equiv \bot$ |

````python
# ============================================================
# Финальный тест по логике высказываний (ipywidgets)
# ============================================================
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

QUESTIONS = [
    ("1. Чему равно ¬¬p?",
     ["p", "¬p", "T всегда", "F всегда"], 0),

    ("2. Чему равно T ∧ F?",
     ["T", "F", "не определено", "зависит от p"], 1),

    ("3. Чему равно F → T?",
     ["T", "F", "не определено", "зависит от p"], 0),

    ("4. Чему равно T ⊕ T?",
     ["T", "F", "T или F", "не определено"], 1),

    ("5. Закон де Моргана: ¬(p ∧ q) = ?",
     ["¬p ∧ ¬q", "¬p ∨ ¬q", "p ∨ q", "¬p ⊕ ¬q"], 1),

    ("6. p → q ≡ ?",
     ["¬p ∧ q", "¬p ∨ q", "p ∧ ¬q", "p ∨ ¬q"], 1),

    ("7. ∀x: P(x) = T если...",
     ["P истинно для большинства x", "P истинно для всех x",
      "P истинно хотя бы для одного x", "P ложно для всех x"], 1),

    ("8. ¬(∀x: P(x)) ≡ ?",
     ["∀x: ¬P(x)", "∃x: ¬P(x)", "∃x: P(x)", "∀x: P(x)"], 1),

    ("9. Что сильнее: ∀x ∃y: P или ∃y ∀x: P?",
     ["∀x ∃y: P", "∃y ∀x: P", "они равны", "зависит от P"], 1),

    ("10. В Python: 'not any(P(x) for x in D)' это то же что...",
     ["all(P(x) for x in D)", "all(not P(x) for x in D)",
      "any(not P(x) for x in D)", "not all(P(x) for x in D)"], 1),
]

score = [0]
outputs = []

for q_text, options, correct_idx in QUESTIONS:
    out = widgets.Output()
    radio = widgets.RadioButtons(options=options, description='', disabled=False)
    btn = widgets.Button(description="Проверить", button_style='info')
    feedback = widgets.HTML(value="")

    def make_check(q=q_text, r=radio, c=correct_idx, f=feedback, o=out, btn=btn):
        def on_click(b):
            with o:
                clear_output()
                chosen = r.index
                if chosen == c:
                    f.value = "<span style='color:green;font-weight:bold'>✓ Верно!</span>"
                    if not getattr(b, '_checked', False):
                        score[0] += 1
                        b._checked = True
                else:
                    correct_letter = "абвг"[c]
                    f.value = (f"<span style='color:red;font-weight:bold'>✗ Неверно.</span> "
                               f"Правильный ответ: <b>{correct_letter})</b>")
                print()
                display(f)
        return on_click

    btn.on_click(make_check())
    outputs.append(widgets.VBox([widgets.HTML(f"<b>{q_text}</b>"), radio, btn, out]))

result_btn = widgets.Button(description="Показать результат", button_style='success')
result_out = widgets.Output()

def show_result(b):
    with result_out:
        clear_output()
        pct = score[0] / len(QUESTIONS) * 100
        emoji = "🏆" if score[0] == 10 else ("🎉" if score[0] >= 8 else ("👍" if score[0] >= 6 else "📚"))
        print(f"{emoji} Твой результат: {score[0]} из {len(QUESTIONS)} ({pct:.0f}%)")
        if score[0] == 10:
            print("\nПрофессор Логичный: 'Безупречно! Можешь спорить с кем угодно — и победить.'")
        elif score[0] >= 8:
            print("\nОтлично! Профессор доволен.")
        elif score[0] >= 6:
            print("\nХорошо! Перечитай темы, где ошибся.")
        else:
            print("\nПрофессор Логичный: 'Логика — не врождённый дар. Учись.'")

result_btn.on_click(show_result)

display(widgets.HTML("<h3>📝 Финальный тест: Логика высказываний</h3>"))
for w in outputs:
    display(w)
display(result_btn, result_out)
````
